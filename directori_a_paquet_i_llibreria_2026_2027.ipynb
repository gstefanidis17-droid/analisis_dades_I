Un <strong>paquete</strong> permite agrupar módulos relacionados bajo un nombre común. En este cuaderno construiremos un <strong>paquete regular</strong>, organizado en una carpeta con un archivo <code>__init__.py</code> y varios archivos <code>.py</code>.

Partiremos de una función sencilla y veremos cómo utilizarla desde otro archivo, ampliar la estructura con un subpaquete y elegir una forma de importación adecuada. El mismo esquema puede servir después para reunir funciones de lectura de datos, cálculos meteorológicos o creación de gráficos.

Una **biblioteca**, también llamada **librería**, reúne recursos reutilizables y puede organizarse como uno o varios paquetes. No es un tipo adicional de carpeta. Aquí trabajaremos con un paquete local, sin instalarlo ni publicarlo.

Los bloques de código que indican un nombre de archivo muestran el contenido que debe guardarse en ese archivo. **Mostrarlos en una celda Markdown no crea los archivos**. El ZIP que acompaña al notebook incluye el proyecto completo con la versión final de los ejemplos.

## El archivo `__init__.py`

En la estructura que vamos a crear, `__init__.py` identifica la carpeta como un **paquete regular**. Puede estar vacío o contener instrucciones de inicialización. Cuando Python carga el paquete por primera vez en una sesión, ejecuta ese archivo; las importaciones posteriores normalmente reutilizan el módulo que ya está en memoria.

De momento, lo dejaremos vacío. **Importar el paquete no carga automáticamente todos sus submódulos**. Más adelante importaremos de forma explícita el módulo o la función que necesitemos.

### Crear un paquete

1. Crea una carpeta llamada `paquete_nuevo` dentro del directorio de tu proyecto.
2. Añade en ella un archivo vacío llamado `__init__.py`, con dos guiones bajos a cada lado de `init`.
3. Guarda en esa misma carpeta los módulos que quieras agrupar, como `mi_modulo.py`.

Comprueba que los archivos tengan realmente la extensión `.py`, no `.py.txt`. El nombre `paquete_nuevo` utiliza letras y un guion bajo, sin espacios ni guiones medios, para poder escribirlo directamente en una importación.

<div style="padding: 0.8em 1em; margin: 0.9em 0 1.1em; border-left: 4px solid #C58A32; background-color: rgba(197,138,50,0.09); border-radius: 4px;">
<strong>Un matiz de Python.</strong> También existen paquetes de espacio de nombres, que no necesitan <code>__init__.py</code>. No los utilizaremos en este ejemplo. Por eso, la regla de añadir este archivo se refiere a los <strong>paquetes regulares</strong>, no a todos los paquetes posibles.
</div>

<p style="font-size: 0.9em; margin: 0.8em 0 0.3em;"><a href="https://docs.python.org/3/reference/import.html#packages">Python · Paquetes regulares y de espacio de nombres</a></p>


### Ejemplo de estructura de un paquete

El directorio `proyecto` contiene nuestra carpeta de trabajo. Dentro de ella, `paquete_nuevo` agrupa los archivos que forman el paquete.

```text
proyecto/
    paquete_nuevo/
        __init__.py
        mi_modulo.py
```

`proyecto` no necesita un `__init__.py` para esta organización, porque no lo utilizaremos como un paquete. El que importaremos se llama `paquete_nuevo`.

### Contenido de `mi_modulo.py`

Guardamos esta función en `paquete_nuevo/mi_modulo.py`.

```python
# paquete_nuevo/mi_modulo.py

def saludar() -> str:
    """Devuelve un saludo sin imprimirlo."""
    return "¡Hola desde mi módulo!"
```

La función devuelve una cadena de texto mediante `return`, pero no la muestra en pantalla. Para verla utilizaremos `print()` desde el programa que la llame.

La anotación `-> str` indica que el resultado esperado es texto. Sirve como información para quien lee el código y para las herramientas de análisis; no convierte el resultado ni comprueba su tipo automáticamente.

<div style="padding: 0.8em 1em; margin: 0.9em 0 1.1em; border-left: 4px solid #59A14F; background-color: rgba(89,161,79,0.08); border-radius: 4px;">
<strong>Organización e importación.</strong> Crear los archivos define la estructura del paquete. Para utilizarlo, Python también debe poder encontrar la carpeta que lo contiene.
</div>

<p style="font-size: 0.9em; margin: 0.8em 0 0.3em;"><a href="https://docs.python.org/3/library/typing.html">Python · Anotaciones de tipo</a></p>


## Importar módulos desde un paquete

Añadimos un archivo `main.py` en el directorio del proyecto, al mismo nivel que `paquete_nuevo`. Ese archivo utilizará la función sin tener que copiar su definición.

```text
proyecto/
    main.py
    paquete_nuevo/
        __init__.py
        mi_modulo.py
```

### Ejemplo de importación

```python
# main.py, situado fuera de paquete_nuevo
from paquete_nuevo.mi_modulo import saludar

print(saludar())
```

En `paquete_nuevo.mi_modulo`, el primer nombre identifica el paquete y el segundo, el módulo. El punto separa ambos niveles y **no escribimos la extensión `.py`**. La parte `import saludar` incorpora la función al archivo actual, de modo que podemos llamarla como `saludar()`.

Desde una terminal situada en `proyecto`, ejecutamos el programa con el intérprete del entorno de trabajo.

```bash
python main.py
```

La salida es la siguiente.

```text
¡Hola desde mi módulo!
```

<div style="padding: 0.8em 1em; margin: 0.9em 0 1.1em; border-left: 4px solid #59A14F; background-color: rgba(89,161,79,0.08); border-radius: 4px;">
<strong>Dónde busca Python.</strong> La carpeta que contiene <code>paquete_nuevo</code>, aquí <code>proyecto</code>, debe estar en la ruta de búsqueda de módulos, <code>sys.path</code>. En la ejecución habitual de <code>main.py</code>, Python incluye la carpeta de ese script. En Jupyter, conviene comprobar el <strong>directorio de trabajo del kernel</strong>, que no tiene por qué coincidir con la carpeta donde está guardado el notebook.
</div>

Para practicar la misma importación desde Jupyter, sitúa el directorio de trabajo en `proyecto`. Puedes consultar su ubicación con este fragmento en una celda de código.

```python
from pathlib import Path

print(Path.cwd())
```

Si aparece `ModuleNotFoundError` al importar `paquete_nuevo`, comprueba primero su ubicación y el nombre de los archivos. Añadir otro `__init__.py` no resuelve por sí solo que Python esté buscando en una carpeta diferente.

<p style="font-size: 0.9em; margin: 0.8em 0 0.3em;"><a href="https://docs.python.org/3/tutorial/modules.html#the-module-search-path">Python · Ruta de búsqueda de módulos</a></p>


## Paquetes con subpaquetes

Un **subpaquete** es un paquete situado dentro de otro. Permite agrupar módulos que comparten una función, por ejemplo, separar la lectura de datos de los cálculos y la visualización.

### Ejemplo de estructura con subpaquetes

Ampliamos el proyecto añadiendo `subpaquete`, también como paquete regular, con su propio `__init__.py` vacío.

```text
proyecto/
    main.py
    paquete_nuevo/
        __init__.py
        mi_modulo.py
        subpaquete/
            __init__.py
            modulo_sub.py
```

### Contenido de `modulo_sub.py`

```python
# paquete_nuevo/subpaquete/modulo_sub.py

def sub_funcion() -> str:
    """Devuelve el saludo del subpaquete."""
    return "¡Hola desde el subpaquete!"
```

### Importación desde un subpaquete

Para probar esta segunda función, sustituimos temporalmente el contenido de `main.py` por el siguiente fragmento.

```python
# main.py
from paquete_nuevo.subpaquete.modulo_sub import sub_funcion

print(sub_funcion())
```

La importación recorre los tres niveles `paquete_nuevo`, `subpaquete` y `modulo_sub`. Al ejecutar de nuevo `main.py`, obtenemos el saludo del subpaquete.

```text
¡Hola desde el subpaquete!
```

La jerarquía de carpetas se refleja en los nombres separados por puntos. No necesitamos cambiar a la carpeta del subpaquete para utilizar sus funciones.

<p style="font-size: 0.9em; margin: 0.8em 0 0.3em;"><a href="https://docs.python.org/3/tutorial/modules.html#packages">Python · Paquetes y subpaquetes</a></p>


## Importación absoluta y relativa

Dentro de un paquete podemos referirnos a otro módulo mediante su nombre completo o mediante su posición en la jerarquía del paquete. Ambas formas permiten acceder al mismo código.

### Importación absoluta

La importación absoluta empieza por el **nombre del paquete de nivel superior**. En este contexto, «absoluta» no significa una ruta del sistema de archivos, como `/home/...` o `C:\...`.

```python
# Dentro de paquete_nuevo/mi_modulo.py
from paquete_nuevo.subpaquete.modulo_sub import sub_funcion
```

La misma instrucción puede utilizarse en `main.py` o en una celda de código, siempre que Python pueda encontrar `paquete_nuevo`.

### Importación relativa

Dentro de `paquete_nuevo/mi_modulo.py`, podemos expresar la misma relación con un punto inicial.

```python
# Dentro de paquete_nuevo/mi_modulo.py
from .subpaquete.modulo_sub import sub_funcion
```

Aquí `.` representa el **paquete al que pertenece el módulo**, `paquete_nuevo`. No representa el directorio de trabajo. Dos puntos iniciales, `..`, permiten subir un nivel en la jerarquía de paquetes, sin salir del paquete de nivel superior.

Para comprobar su utilidad, ampliamos `mi_modulo.py` con una función que reúne los dos saludos. Esta es su **versión completa**, que sustituye la anterior y mantiene `saludar()` sin cambiar su resultado.

```python
# paquete_nuevo/mi_modulo.py
from .subpaquete.modulo_sub import sub_funcion


def saludar() -> str:
    """Devuelve un saludo sin imprimirlo."""
    return "¡Hola desde mi módulo!"


def saludar_desde_ambos() -> str:
    """Combina los saludos del módulo y del subpaquete."""
    return f"{saludar()}\n{sub_funcion()}"
```

`saludar_desde_ambos()` llama a una función del propio módulo y a otra importada desde el subpaquete. La secuencia `\n` introduce un salto de línea entre los mensajes.

<div style="padding: 0.8em 1em; margin: 0.9em 0 1.1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px;">
<strong>Elegir una forma de importación.</strong> Las importaciones absolutas hacen visible el origen del código y son la recomendación general de PEP 8. Las relativas explícitas también son válidas para las relaciones internas de un paquete. No son, por sí mismas, más difíciles de mantener. Conviene elegir una forma coherente con la estructura del proyecto.
</div>

Las importaciones relativas necesitan un **contexto de paquete**. Por eso, una instrucción que empieza por `.` no se utiliza tal cual en una celda de notebook ni en el `main.py` externo de este ejemplo.

Tampoco ejecutamos `paquete_nuevo/mi_modulo.py` como un archivo suelto, porque perdería ese contexto. Desde `proyecto`, la opción `-m` permite ejecutarlo por su nombre de módulo, sin añadir `.py`.

```bash
python -m paquete_nuevo.mi_modulo
```

Esta orden carga el módulo dentro del contexto de su paquete y permite resolver la importación relativa. **No muestra ningún saludo**, porque el archivo solo importa una función y define otras. Para obtener la salida, las llamaremos desde `main.py`.

<p style="font-size: 0.9em; margin: 0.8em 0 0.3em;"><a href="https://docs.python.org/3/reference/import.html#package-relative-imports">Python · Importaciones relativas</a> · <a href="https://peps.python.org/pep-0008/#imports">PEP 8 · Importaciones</a> · <a href="https://docs.python.org/3/using/cmdline.html#cmdoption-m">Python · Ejecución con -m</a></p>


## El archivo `main.py` y su relación con los paquetes

`main.py` es un nombre convencional para el **punto de entrada**, el archivo desde el que ponemos en marcha una aplicación. No tiene un significado especial para Python y no es necesario para que `paquete_nuevo` sea un paquete.

Separamos así el código reutilizable, que permanece en el paquete, de las instrucciones que ejecutan una tarea concreta. La versión final de `main.py` utiliza la función que acabamos de añadir.

```python
# main.py, situado junto a la carpeta paquete_nuevo
from paquete_nuevo.mi_modulo import saludar_desde_ambos


def main() -> None:
    """Muestra los dos saludos usando el código del paquete."""
    print(saludar_desde_ambos())


# Esta llamada se realiza al ejecutar el archivo, no al importarlo.
if __name__ == "__main__":
    main()
```

La función `main()` organiza la tarea del programa. La anotación `-> None` indica que no se espera un valor de retorno útil, porque aquí la función se limita a mostrar el texto.

La condición `if __name__ == "__main__":` distingue dos situaciones. Cuando ejecutamos `main.py` como programa, `__name__` toma el valor `"__main__"` y se llama a `main()`. Si otro archivo hace `import main`, se cargan sus importaciones y definiciones, pero **no se ejecuta esa llamada**.

Desde `proyecto`, la misma orden de terminal ejecuta ahora el ejemplo completo.

```bash
python main.py
```

```text
¡Hola desde mi módulo!
¡Hola desde el subpaquete!
```

Estos tres nombres se parecen, pero cumplen funciones distintas.

| Archivo | Papel en el proyecto |
|:--|:--|
| `__init__.py` | Inicializa un paquete regular al cargarlo. Puede estar vacío. |
| `main.py` | Es un nombre convencional para el programa principal. Aquí está fuera del paquete. |
| `__main__.py` | Permite ejecutar el propio paquete con `python -m paquete_nuevo`. Es opcional y no se incluye en este ejemplo. |
