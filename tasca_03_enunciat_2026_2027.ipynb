## 2. Elegir qué observaciones mirar

Escoge Litoral o Interior y explica brevemente qué te interesa explorar en ella. Construye una máscara sobre `Estacion` y guarda las filas elegidas en `estacion`, utilizando `loc`.

```python
#Escojo litoral por ejemplo para poder analizar las brisas marinas.

# Construimos la máscara para filtrar la estación "Litoral"
mascara_litoral = datos["Estacion"] == "Litoral"
# Guardamos las filas elegidas en la variable 'estacion' usando .loc
estacion = datos.loc[mascara_litoral]
# Mostramos las primeras filas de estacion
display(estacion.head())
# Mostramos el número total de filas
print("Número de filas:", len(estacion))
```

Crea `serie` a partir de `estacion`, usando `Fecha` como índice y ordenándolo cronológicamente. Puedes encadenar `set_index()` y `sort_index()`.

Muestra sus dos primeras filas y consulta `serie.index.is_unique`. Explica por qué ahora una fecha puede identificar un único registro.

```python
serie = estacion.set_index('Fecha').sort_index()
display(serie.head(2))
serie.index.is_unique
```
