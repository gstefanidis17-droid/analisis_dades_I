Me propongo calcular CDD (cooling degree-days) de los datos "serie". Los CDD son un indicador de demanda energética ligado al uso de sistemas de climatización como el A/C que cuantifica el augmento de la demanda de energia debido a las altas temperaturas. Funciona como una especie de contador que augmenta su valor segun el numero de días donde se haya superado un cierto umbral de temperatura. Típicamente entre abril y septiembre podemos encontrarnos con temperaturas altas que impliquen un cierto nivel de CDD.
Existen diversas formas de determinar estos CDD, algunas propuestas de cálculo requieren de temperaturas mínimas, máximas y medias diarias, otros autores (J.Spinoni et al.) proponen métodos para calcularlo que solo implican la temperatura media mensual. En mi caso solo dispongo de la temperatura media diaria así que voy a crear mi propio indicador simple (sin demasiada base física, simplemente para hacer el ejercicio).
Marcaré mi temperatura umbral en 22ºC, si la temperatura media del dia es estrictamente inferior no sumaremos nada al contador de CDD. Si la temperatura media es igual o superior al umbral pero inferior al umbral más un grado sumaremos 1 al contador de CDD. Finalmente si la temperatura es igual o superior al umbral más un grado, sumaremos 2 al contador de CDD.
Mi objetivo con esto es determinar si los días de "serie" han podido suponer un incremento significativo en la demanda energética (para la época del año) y como consecuencia, una sobrecarga en la red.

```python
# Defino mi función de indicador CDD que me da el CDD del dia
def CDD(temp_dia, temp_umbral):
    if temp_dia < temp_umbral:
        return 0
    elif temp_dia >= temp_umbral + 1:
        return 2
    else:
        return 1
# Ahora aplico mi función a los datos de serie y hago el sumatorio del total de CDD de todos los días
temp_umbral = 22  # Umbral

# Aplicamos la función a la columna 'Temperatura_C' y guardamos los resultados en una nueva columna llamada CDD
serie['CDD'] = serie['Temperatura_C'].apply(lambda x: CDD(x, temp_umbral))
# Sumamos el total del contador
contador_total = serie['CDD'].sum()

print(f"Contador total de CDD: {contador_total}")
 
```

Podemos ver que el valor de CDD es de 2, un valor bajo tratandose de 14 días. Por lo tanto podemos determinar que no fue un mes de mayo especialmente caluroso, con lo que tampoco se llego a incrementar mucho la demanda energética a causa de la climatización de los hogares.
