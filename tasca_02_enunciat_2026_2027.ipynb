## 2. Calcular las medias 

Define `media_temperaturas(valores)` con un bucle `for`: debe devolver la media de los valores y devolver `None` si la serie está vacía.

Antes de ejecutar los cálculos, anticipa el resultado para cada una de estas series. Después llama a la función con ellas y contrasta tu previsión.

| Caso | Serie de temperaturas (°C) |
|:--|:--|

| B | `[0.0]` |
| C | `[]` |

Explica qué significa cada resultado y por qué una serie vacía no debe producir una media de cero. Distingue el parámetro de la función de los argumentos utilizados al llamarla.

```python
# Lo esperable es que retorne una media de 24 y luego una de 0. Para que no devuelva un 0 también para la tercedra 
# lista lo tenemos que forzar con condiciones. Esto es necesario para detectar lista sin información y diferenciar-
#las de listas de media realmente igual a 0.
def media_temperaturas(valores):
    contador=0
    suma=0
    media=0
    if  len(valores) == 0:
        return "Lista sin valores"
    
    else:
        
        for i in valores:
            contador=contador+1
            suma=suma+i
        media=suma/contador
    return media

# Vamos a ver si funciona la función
A = [18.0, 24.0, 30.0]
B = [0.0]
C = []

print("Media lista A:", media_temperaturas(A))
print("Media lista B:", media_temperaturas(B))
print("Media lista C:", media_temperaturas(C))
```

Escribe dos funciones equivalentes, **`contar_for(valores, umbral)`** y **`contar_while(valores, umbral)`**, que devuelvan cuántas temperaturas superan estrictamente el umbral. Usa un bucle `for` en la primera y `while` en la segunda.

Llama a ambas con la serie y el umbral de S03, tomados de la estructura anterior. Muestra los resultados y comprueba con una comparación que coinciden. Explica qué hace avanzar el `while` y por qué termina.

```python
def contar_for(valores, umbral):
    contador=0
    for valor in valores:
        if valor > umbral:
            contador += 1
    return contador

def contar_while(valores, umbral):
    contador=0
    n=0
    while n < len(valores):
        valor = valores[n]
        if valor > umbral:
            contador += 1
        n += 1
    return contador 
    
temp_S03 = datos_dic["S03"]["temperaturas"]
umbral_S03 = datos_dic["S03"]["umbral"]

print(contar_for(temp_S03, umbral_S03))
print(contar_while(temp_S03, umbral_S03))

# While compara el valor de n con la longitud de valores (en mi caso es 4) y va iterando hasta que n ya no es
# estrictamente menor a len(valores).
```
