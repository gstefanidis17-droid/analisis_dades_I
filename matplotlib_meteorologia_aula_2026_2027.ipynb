### 5. Bandas y correlaciones tienen interpretaciones diferentes

<div style="padding: 0.8em 1em; margin: 0.8em 0 1.2em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">La banda de media móvil ± una desviación estándar describe la variabilidad de los datos de cada ventana. En una dispersión, la recta y el coeficiente <code>r</code> resumen una asociación lineal entre pares. Ambas representaciones necesitan una interpretación ajustada al cálculo realizado y a su muestra.</p>
</div>

<div style="padding: 0.8em 1em; margin: 0.8em 0 1.2em; border-left: 4px solid #B88632; background-color: rgba(184,134,50,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0 0 0.4em; font-weight: 600;">Antes de continuar</p><p style="margin: 0.45em 0;">¿La banda describe incertidumbre? ¿Un r negativo demuestra causalidad?</p>
</div>

<details style="padding: 0.8em 1em; margin: 0.8em 0 1.2em; border-left: 4px solid #59A14F; background-color: rgba(89,161,79,0.08); border-radius: 4px; line-height: 1.6;">
<summary style="cursor: pointer; font-weight: 600;">Respuesta</summary>
<div style="margin-top: 0.65em;">
<p style="margin: 0.45em 0;">La desviación local describe dispersión, no incertidumbre de la media. En los extremos, una ventana centrada puede contener menos datos. Un <code>r</code> negativo indica asociación lineal decreciente, sin demostrar causalidad ni una tendencia temporal. Nube de puntos, ajuste y correlación deben usar los mismos pares válidos; la recta se dibuja dentro del rango observado.</p>
</div>
</details>

### 7. El color distingue cantidades y ausencias

<div style="padding: 0.8em 1em; margin: 0.8em 0 1.2em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">

El cero es un acumulado válido; una celda sin datos necesita una representación identificable que no se confunda con él.</p>
