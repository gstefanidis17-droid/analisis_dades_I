### Ejercició 10: número de días entre fechas

Implementa una función llamada `dias_entre(d1, d2) donde d1 y d2 son cadenas de texto con el formato "YYYY-MM-DD".
La función debe devolver el número de días que hay entre ambas fechas, utilizando el módulo datetime de Python.

````python
from datetime import datetime
def dias_entre(d1, d2):
    fecha_1 = datetime.strptime(d1, "%Y-%m-%d")
    fecha_2 = datetime.strptime(d2, "%Y-%m-%d")
    resultado = (fecha_2 - fecha_1).days
    return resultado
# Ejecutamos un ejemplo
d1="2023-01-13"
d2="2026-10-02"
dias_entre(d1, d2)
````

### Ejercicio 11: creación de clases

Crea una clase llamada EstacionMeteorologica con los siguientes atributos:

* Ubicación
* Latitud
* Longitud
* Altitud

La clase debe tener los siguientes métodos:

* registrar_temperatura(self, temperatura, fecha): Registra una temperatura en una fecha específica.
* obtener_temperatura_promedio(self, fecha_inicio, fecha_fin): Calcula la temperatura promedio entre dos fechas.
*registrar_precipitacion(self, precipitacion, fecha): Registra una precipitación en una fecha específica.
*obtener_precipitacion_total(self, fecha_inicio, fecha_fin): Calcula la precipitación total entre dos fechas.

Crea una instancia de la clase y demuestra su funcionamiento.

````python
import datetime

#---------CLASE----------------------------------------------------------------------
class Estacion_Meteorologica:
# La función se ejecuta al crear la estación meteorologica y fija sus datos iniciales
    def __init__(self, ubicacion, latitud, longitud, altitud):
        self.ubicacion = ubicacion
        self.latitud = latitud
        self.longitud = longitud
        self.altitud = altitud
        # Diccionarios para guardar los datos 
        self.temperaturas = {} 
        self.precipitaciones = {}
# Registro temperatura de la fecha y me permite escribirla en temperaturas
    def registrar_temperatura(self, temperatura, fecha): 
        self.temperaturas[fecha] = temperatura
#Calculo la temperatura promedio
    def obtener_temperatura_promedio(self, fecha_inicio, fecha_fin):
        temperaturas = []

        for fecha, temperatura in self.temperaturas.items():
            if fecha_inicio <= fecha <= fecha_fin:
                temperaturas.append(temperatura)

        if len(temperaturas) == 0:
            return None

        return sum(temperaturas) / len(temperaturas)
# Registrar la precipitación
    def registrar_precipitacion(self, precipitacion, fecha):
        self.precipitaciones[fecha] = precipitacion
# Calculo la precipitación total entre fechas
    def obtener_precipitacion_total(self, fecha_inicio, fecha_fin):
        total = 0
        for fecha, precipitacion in self.precipitaciones.items():
            if fecha_inicio <= fecha <= fecha_fin:
                total += precipitacion
        return total
#-------CLASE CREADA-------------------------------------------------------------

# Ahora la pondre a prueba con algún ejemplo

# Crear una instancia de la clase 
estacion = Estacion_Meteorologica( "Barcelona", 41.3874, 2.1686, 12 )
# Registrar temperaturas 
estacion.registrar_temperatura(20, "2026-09-15") 
estacion.registrar_temperatura(22, "2026-09-16") 
estacion.registrar_temperatura(19, "2026-09-17") 
estacion.registrar_temperatura(21, "2026-09-18")
# Registrar precipitaciones 
estacion.registrar_precipitacion(5, "2026-09-15") 
estacion.registrar_precipitacion(0, "2026-09-16") 
estacion.registrar_precipitacion(10, "2026-09-17") 
estacion.registrar_precipitacion(3, "2026-09-18")
# Obtener temperatura promedio 
promedio = estacion.obtener_temperatura_promedio( "2026-09-15", "2026-09-18" ) 
print("Temperatura promedio:", promedio, "°C") 
# Obtener precipitación total 
total = estacion.obtener_precipitacion_total( "2026-09-15", "2026-09-18" ) 
print("Precipitación total:", total, "mm")
````
