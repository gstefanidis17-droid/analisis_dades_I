### 2. Etiquetas, posiciones y condiciones seleccionan de formas distintas

<div style="padding: 0.8em 1em; margin: 1em 0; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">

Un corte temporal con <code>loc</code> incluye ambos extremos en nuestro índice ordenado. Una máscara selecciona los registros que cumplen una condición. Combinamos comparaciones con <code>&amp;</code> o <code>|</code>, colocando cada comparación entre paréntesis.</p>
</div>

## Caso práctico · Preparar un resumen de dos estaciones

<div style="height: 4px; background-color: #4C78A8; border-radius: 2px; margin: 0.8em 0 1.2em;"></div>

Recibimos dos tablas diarias de las estaciones ficticias **COSTA** e **INTERIOR**, del 1 de enero al 28 de febrero de 2026. Queremos preparar un resumen mensual de temperatura y precipitación, conocer la disponibilidad de cada variable y comparar las temperaturas de los mismos días.

La celda siguiente construye los datos sintéticos de partida. Hay una fila por estación y día, con algunas medidas ausentes. La temperatura y la humedad son medias diarias; la precipitación es un acumulado diario. Cada día corresponde al intervalo completo de 00 a 24 UTC.

| Columna | Contenido | Unidad o formato |
| --- | --- | --- |
| `Fecha` | Etiqueta del día, recibida como texto | `dd/mm/aaaa` |
| `Temperatura_C` | Temperatura media diaria | °C |
| `Humedad` | Humedad relativa media diaria | Fracción entre 0 y 1 |
| `Precipitacion_mm` | Precipitación acumulada diaria | mm |

<div style="padding: 0.8em 1em; margin: 1em 0; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.35em 0;"><strong>Datos de partida</strong></p><p style="margin: 0.35em 0;">En <code>Precipitacion_mm</code>, el código <code>-999</code> significa dato ausente. Los valores <code>0.0</code> mm son válidos. Otras ausencias ya aparecen como <code>NaN</code>. Los metadatos también son ficticios y contienen una sola fila por estación.</p><p style="margin: 0.35em 0;">Ejecutad la preparación y resolved los apartados en orden. Conservad <code>registros_costa</code>, <code>registros_interior</code> y <code>metadatos</code> como tablas de partida.</p>
</div>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

plt.rcParams.update({
    "figure.dpi": 120, "axes.labelsize": 11,
    "axes.titlesize": 12, "legend.fontsize": 10,
})

calendario = pd.date_range("2026-01-01", "2026-02-28", freq="D")


def crear_registros(semilla, temperatura_base_C):
    """Construye una tabla sintética con fechas de entrada como texto."""
    rng = np.random.default_rng(semilla)
    dias = np.arange(len(calendario))
    tabla = pd.DataFrame({
        "Fecha": calendario.strftime("%d/%m/%Y"),
        "Temperatura_C": np.round(
            temperatura_base_C + 3 * np.sin(dias / 6)
            + rng.normal(0, 0.5, len(dias)), 1
        ),
        "Humedad": np.round(rng.uniform(0.45, 0.95, len(dias)), 2),
        "Precipitacion_mm": rng.choice(
            [0.0, 0.0, 0.0, 0.0, 0.5, 2.0, 6.0, 15.0, 25.0], len(dias)
        ),
    })
    # Los registros llegan en orden cronológico inverso.
    return tabla.iloc[::-1].reset_index(drop=True)


registros_costa = crear_registros(17, 12.0)
registros_interior = crear_registros(29, 7.0)

registros_costa.loc[
    registros_costa["Fecha"].isin(["10/01/2026", "05/02/2026"]), "Temperatura_C"
] = np.nan
registros_interior.loc[
    registros_interior["Fecha"].isin(["12/01/2026", "23/01/2026", "12/02/2026"]),
    "Temperatura_C"
] = np.nan
registros_costa.loc[registros_costa["Fecha"] == "18/01/2026", "Humedad"] = np.nan
registros_interior.loc[registros_interior["Fecha"] == "16/02/2026", "Humedad"] = np.nan

# En la documentación de esta práctica, -999 identifica una precipitación ausente.
registros_costa.loc[
    registros_costa["Fecha"].isin(["03/02/2026", "18/02/2026"]), "Precipitacion_mm"
] = -999.0
fechas_febrero = calendario[calendario.month == 2].strftime("%d/%m/%Y")
registros_interior.loc[
    registros_interior["Fecha"].isin(fechas_febrero), "Precipitacion_mm"
] = np.nan

# Metadatos ficticios, en un orden distinto al de las tablas anteriores.
metadatos = pd.DataFrame({
    "Estacion": ["INTERIOR", "COSTA"],
    "ID_estacion": [202, 101],
    "Entorno": ["Interior", "Litoral"],
    "Altitud_m": [450.0, 20.0],
})
```
