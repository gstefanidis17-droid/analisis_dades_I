## Concepto de rutas

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Una <strong>ruta</strong>, o <em>path</em> en inglés, describe cómo localizar un archivo o un directorio dentro del sistema de archivos. Un <strong>directorio</strong> es lo que habitualmente llamamos una carpeta. Podemos imaginar la ruta como una dirección que indica dónde buscar un archivo o dónde queremos guardarlo.</p>
</div>

En Python podemos representar las rutas mediante cadenas de texto o con objetos específicos, como los que proporciona `pathlib`. La distinción fundamental es si la ubicación se describe desde la raíz del sistema de archivos o desde un punto de partida cercano.

- **Ruta absoluta**. Indica la ubicación completa a partir de la raíz y, en Windows, incluye también la unidad o el recurso de red correspondiente. No depende del directorio de trabajo. Por ejemplo, `/home/usuario/documentos/archivo.txt` es una ruta absoluta en Linux, mientras que `C:\Users\usuario\documentos\archivo.txt` lo es en Windows.

- **Ruta relativa**. Describe cómo llegar al archivo desde el **directorio de trabajo**, que es la carpeta que el programa utiliza como referencia. Si este directorio es `/home/usuario`, la ruta `documentos/archivo.txt` apunta a `/home/usuario/documentos/archivo.txt`. También podemos escribirla como `./documentos/archivo.txt`.

Las rutas relativas facilitan trasladar un proyecto a otro ordenador, siempre que se mantengan la organización de sus carpetas y el punto de partida desde el que se interpretan.

### Breve apunte sobre rutas relativas

El directorio de trabajo es la referencia desde la que Python interpreta las rutas relativas. Por ejemplo, `./data/mi_archivo.txt` apunta a un archivo de la carpeta `data` situada dentro de ese directorio.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #C28C38; background-color: rgba(194,140,56,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">En Jupyter, el <strong>directorio de trabajo no tiene por qué coincidir con la carpeta que contiene el notebook</strong>. La referencia es la del <em>kernel</em>, el proceso que ejecuta el código, y puede cambiar durante la sesión. Conviene comprobarla en lugar de deducirla a partir de la ubicación del archivo <code>.ipynb</code>.</p>
</div>

Podemos consultar esta ubicación con `Path.cwd()`. La abreviatura *cwd* procede de *current working directory*.

```python
from pathlib import Path

print(Path.cwd())
```

En las celdas de código de Jupyter podemos usar `%pwd` para consultar el directorio y `%cd ..` para subir un nivel. Son órdenes de IPython, no instrucciones ordinarias de Python. La alternativa en Python es `os.chdir()`, que veremos más adelante. Usar `!cd` no cambia de forma persistente el directorio del kernel, porque esa orden se ejecuta en un proceso separado.

La documentación de [las órdenes de IPython](https://ipython.readthedocs.io/en/stable/interactive/magics.html#magic-cd) describe estas operaciones con más detalle.

Veamos cómo se interpretan estas rutas en un proyecto con una carpeta para los notebooks y otra para los datos.

```text
proyecto/
├── notebooks/
│   └── mi_notebook.ipynb
└── data/
    └── mi_archivo.txt
```

Si el directorio de trabajo es `proyecto/notebooks`, la ruta `../data/mi_archivo.txt` sube primero a `proyecto` y después entra en `data`. Si trabajamos desde `proyecto`, basta con `data/mi_archivo.txt`.

<p style="margin: 0.45em 0;">Una misma ruta relativa puede apuntar a ubicaciones distintas si cambia el directorio de trabajo. Para interpretarla, necesitamos conocer tanto la ruta como su punto de partida.</p>

## Importancia de la gestión de rutas

Una gestión cuidadosa de las rutas ayuda a que el programa encuentre los datos de entrada y guarde los resultados donde corresponde. Si el punto de partida o el nombre del archivo no son los esperados, podemos obtener un error de archivo no encontrado o trabajar con un archivo distinto del que pretendíamos utilizar.

Construir bien una ruta no garantiza que el archivo exista ni que tengamos permiso para leerlo o modificarlo. También debemos comprobar qué operación realizamos, especialmente al escribir o eliminar archivos, para evitar pérdidas de información.

Al escribir una ruta de Windows dentro de una cadena de Python, la barra inversa `\` requiere atención porque también introduce secuencias especiales, como `\n` para un salto de línea o `\t` para una tabulación. Podemos usar una **cadena sin procesar**, precedida de `r`, o duplicar cada barra inversa.

```python

ruta_windows = r"C:\Users\usuario\documentos\archivo.txt"
ruta_windows_equivalente = "C:\\Users\\usuario\\documentos\\archivo.txt"
```

En Python sobre Windows también se admite `/` en muchas operaciones con archivos. Aun así, suele resultar más claro construir las rutas a partir de sus componentes en lugar de escribir los separadores manualmente.

- **Módulo `os`**. Proporciona funciones para interactuar con el sistema operativo. Dentro de él, `os.path.join()` une las partes de una ruta según las convenciones del sistema, y `os.sep` permite consultar su separador habitual.

- **Módulo `pathlib`**. Representa las rutas como objetos. La clase `Path` permite construirlas, consultar sus componentes y realizar operaciones con archivos mediante una sintaxis común en Windows, Linux y macOS.

### Ejemplos prácticos con `os`

El siguiente ejemplo consulta el directorio de trabajo, crea una carpeta llamada `ejemplo_os` y guarda en ella un pequeño archivo de texto. Después comprueba por separado si una ruta corresponde a un archivo o a un directorio.

El modo `"x"` de `open()` crea un archivo únicamente si no existe. Si el destino ya está ocupado, se produce `FileExistsError`, que aquí capturamos para conservar lo que hubiera. El bloque `with` se encarga de cerrar el archivo al terminar, y `encoding="utf-8"` fija la codificación del texto.

````python
import os

directorio_actual = os.getcwd()
print(f"Directorio de trabajo: {directorio_actual}")
print(f"Separador de rutas del sistema: {os.sep}")

# Ordenamos el listado para facilitar su lectura.
print("Elementos del directorio de trabajo")
for nombre in sorted(os.listdir(directorio_actual)):
    print(f"- {nombre}")

# exist_ok=True permite reutilizar la carpeta sin que falle su creación.
carpeta = os.path.join(directorio_actual, "ejemplo_os")
os.makedirs(carpeta, exist_ok=True)

ruta = os.path.join(carpeta, "archivo.txt")
print(f"Ruta construida: {ruta}")
print(f"¿Hay un archivo antes de escribir? {os.path.isfile(ruta)}")

# El modo "x" evita sobrescribir un destino que ya exista.
try:
    with open(ruta, "x", encoding="utf-8") as archivo:
        archivo.write("Este es un archivo de ejemplo.\n")
except FileExistsError:
    print("El destino ya existe y se conserva sin cambios.")

print(f"¿La ruta corresponde a un archivo? {os.path.isfile(ruta)}")
print(f"¿La carpeta es un directorio? {os.path.isdir(carpeta)}")
print(f"¿La ruta del archivo es un directorio? {os.path.isdir(ruta)}")
````

### Manipulación de rutas con `pathlib`

Importamos `Path` y utilizamos el operador `/` para unir los componentes. En este contexto, `/` no realiza una división, sino que construye una nueva ruta.

```python
from pathlib import Path

ruta = Path("directorio") / "subdirectorio" / "archivo.txt"
print(ruta)
```

Al imprimirla, veremos `directorio/subdirectorio/archivo.txt` en Linux y macOS, o `directorio\subdirectorio\archivo.txt` en Windows. El símbolo `/` que escribimos en el código no obliga a que el separador mostrado sea una barra inclinada.

Para comprobar qué hay en esa ubicación podemos usar `ruta.exists()`, `ruta.is_file()` o `ruta.is_dir()`. La primera operación comprueba la existencia; las otras dos permiten distinguir entre un archivo y un directorio.

### Ejemplos prácticos con `pathlib`

Trabajaremos dentro de `ejemplo_pathlib` para mantener separados los archivos de cada ejemplo. Primero intentaremos leer `archivo.txt`. Si todavía no existe, capturaremos `FileNotFoundError` y mostraremos un mensaje que explique lo ocurrido. Después lo crearemos sin sobrescribir un destino existente y consultaremos su contenido.

````python
from pathlib import Path

directorio_actual = Path.cwd()
print(f"Directorio de trabajo: {directorio_actual}")

carpeta = directorio_actual / "ejemplo_pathlib"
carpeta.mkdir(parents=True, exist_ok=True)
ruta = carpeta / "archivo.txt"
print(f"Ruta construida: {ruta}")

# La primera lectura muestra qué ocurre cuando el archivo aún no existe.
try:
    contenido = ruta.read_text(encoding="utf-8")
except FileNotFoundError:
    print("El archivo todavía no existe y no se puede leer.")
else:
    print(f"Contenido previo: {contenido.strip()}")

# Path.open() también admite la creación exclusiva con el modo "x".
try:
    with ruta.open("x", encoding="utf-8") as archivo:
        archivo.write("Este es el contenido del archivo de ejemplo.\n")
except FileExistsError:
    print("El destino ya existe y se conserva sin cambios.")

print(f"¿Es un archivo? {ruta.is_file()}")
print(f"¿Su carpeta es un directorio? {ruta.parent.is_dir()}")
print(f"Nombre del archivo: {ruta.name}")
print(f"Carpeta que lo contiene: {ruta.parent}")

if ruta.is_file():
    contenido = ruta.read_text(encoding="utf-8")
    print(f"Contenido leído: {contenido.strip()}")
````

### Ajustar el comportamiento del programa según el sistema operativo

Cuando una ubicación depende realmente del sistema, podemos seleccionar una ruta mediante un condicional. Las siguientes rutas son ilustrativas y no implican que esas carpetas existan en nuestro equipo.

```python
import platform
from pathlib import Path

sistema = platform.system()

if sistema == "Windows":
    ruta = r"C:\Users\usuario\mi_directorio\archivo.txt"
elif sistema == "Linux":
    ruta = "/home/usuario/mi_directorio/archivo.txt"
elif sistema == "Darwin":
    ruta = "/Users/usuario/mi_directorio/archivo.txt"
else:
    # Ofrecemos una alternativa en lugar de dejar la variable sin definir.
    ruta = str(Path.home() / "mi_directorio" / "archivo.txt")

print(ruta)
```

No necesitamos escribir una rama para cada sistema si solo queremos construir una ruta dentro de la carpeta personal. En ese caso, `Path.home()` proporciona un punto de partida común.

```python
from pathlib import Path

```

Esta última expresión adapta la ruta al usuario que ejecuta Python, pero no crea `mi_directorio` ni `archivo.txt`. La documentación de [`platform`](https://docs.python.org/3/library/platform.html#platform.system) amplía la información sobre la identificación del sistema.

### Recomendaciones generales al trabajar con rutas

#### Evitar rutas fijas dependientes del equipo

Una ruta *hardcoded* está escrita directamente en el código. El problema aparece sobre todo cuando incluye nombres de usuario o carpetas particulares de un ordenador. En cambio, nombres relativos como `data/archivo.txt` pueden formar parte de la estructura prevista de un proyecto.

**Centralizar la ruta base** facilita los cambios y evita repetirla en distintas partes del programa.

```python
import os

directorio_datos = "data"
ruta_archivo = os.path.join(directorio_datos, "archivo.txt")
```

Si cambia el nombre de la carpeta, basta con modificar `directorio_datos`. Guardar una ruta absoluta en una variable facilita su mantenimiento, pero no elimina por sí solo la dependencia del equipo.

**Construir nombres a partir del contexto** permite identificar archivos asociados a un usuario o a una fecha. Por ejemplo, podemos generar el nombre de un archivo de registro, o *log*, con ambos datos.

```python
import getpass
from datetime import datetime
from pathlib import Path

# getuser() no requiere una sesión de terminal interactiva.
usuario = getpass.getuser()
fecha_actual = datetime.now().strftime("%Y-%m-%d")
ruta_log = Path("logs") / f"log_{usuario}_{fecha_actual}.txt"
print(ruta_log)
```

Este código construye una ruta, pero no crea la carpeta `logs` ni escribe el archivo. El nombre tampoco es único para cada ejecución, porque el mismo usuario obtendrá el mismo nombre durante todo el día. Si necesitamos distinguir varias ejecuciones, habrá que incorporar otro identificador y comprobar los posibles conflictos.

**Usar rutas relativas dentro del proyecto** reduce la dependencia de su ubicación en el ordenador.

```python
from pathlib import Path

ruta_relativa = Path("data") / "archivo.txt"
```

Para que esta ruta siga funcionando al trasladar el proyecto, debemos conservar la estructura de carpetas y utilizar un directorio de trabajo coherente con ella.

**Separar la configuración del código** puede resultar útil en proyectos más complejos. Las rutas se pueden guardar en archivos `.ini`, `.json` o `.yaml`, de modo que podamos ajustarlas sin modificar el programa. Por ejemplo, un archivo `config.json` puede contener lo siguiente.

```json
{
  "ruta_archivo": "data/archivo.txt"
}
```

El siguiente fragmento supone que hemos guardado `config.json` en el directorio de trabajo. Interpreta la ruta de los datos respecto a la carpeta de ese archivo de configuración.

```python
import json
from pathlib import Path

ruta_config = Path("config.json").resolve()
with ruta_config.open(encoding="utf-8") as archivo_config:
    configuracion = json.load(archivo_config)

ruta_archivo = ruta_config.parent / configuracion["ruta_archivo"]
print(ruta_archivo)
```

<p style="margin: 0.45em 0;">Una ruta debe tener un punto de partida claro. Centralizar esa referencia y construir a partir de ella las rutas de entrada y salida suele ser más útil que cambiar repetidamente el directorio de trabajo.</p>

## Ejercicio

En este ejercicio practicaremos cómo localizar un archivo en la carpeta actual, en una subcarpeta y en una carpeta situada un nivel por encima.

Toma como directorio de trabajo la carpeta que contiene este notebook y comprueba su ubicación con `os.getcwd()` o `Path.cwd()`. Puedes crear los archivos y hacer las copias desde el explorador de archivos o desde JupyterLab; las comprobaciones se realizan con Python.

1. **Crear un archivo de ejemplo**

3. **Crear una copia en una carpeta del directorio superior**

   Crea una carpeta llamada `copias_superior` en el directorio que contiene al actual. Copia allí `mi_archivo.txt`, comprueba su existencia y muestra su ruta. El destino debe ser `../copias_superior/mi_archivo.txt`, no un archivo colocado directamente en el directorio superior.

Las dos celdas siguientes contienen soluciones alternativas, una con `os` y otra con `pathlib`. Ambas automatizan también la creación y las copias, y comparan el contenido de los archivos para no confundir la mera existencia de un destino con una copia correcta.

En las soluciones, `"rb"` permite leer los bytes del archivo y `"xb"` crea una copia binaria solo si el destino no existe. Se lee el contenido completo porque trabajamos con un archivo de texto pequeño. Cada solución reúne los dos destinos en una lista de pares, con una descripción y una carpeta, y utiliza un bucle para aplicar la misma operación a ambos. La documentación de [`open()`](https://docs.python.org/3/library/functions.html#open) recoge los distintos modos de apertura.

````python
# Solución con os
import os

directorio_actual = os.getcwd()
ruta_original = os.path.join(directorio_actual, "mi_archivo.txt")

print("1. Archivo de ejemplo")
try:
    with open(ruta_original, "x", encoding="utf-8") as archivo:
        archivo.write("Este es un archivo de ejemplo.\n")
except FileExistsError:
    print("El destino ya existe y no se sobrescribe.")

# No basta con que la ruta exista; debe corresponder a un archivo.
if not os.path.isfile(ruta_original):
    raise ValueError(f"La ruta original no corresponde a un archivo: {ruta_original}")

print(f"Archivo original: {ruta_original}")
print("Elementos del directorio de trabajo")
for nombre in sorted(os.listdir(directorio_actual)):
    print(f"- {nombre}")

with open(ruta_original, "rb") as archivo:
    contenido_original = archivo.read()

# Las dos carpetas parten de referencias diferentes.
subcarpeta = os.path.join(directorio_actual, "nueva_carpeta")
directorio_superior = os.path.dirname(directorio_actual)
carpeta_superior = os.path.join(directorio_superior, "copias_superior")

destinos = [
    ("2. Copia en una subcarpeta", subcarpeta),
    ("3. Copia en una carpeta del directorio superior", carpeta_superior),
]

for descripcion, carpeta in destinos:
    print(f"\n{descripcion}")
    os.makedirs(carpeta, exist_ok=True)
    ruta_copia = os.path.join(carpeta, os.path.basename(ruta_original))

    try:
        with open(ruta_copia, "xb") as copia:
            copia.write(contenido_original)
    except FileExistsError:
        print("El destino ya existe y se conserva sin cambios.")

    print(f"Ruta de destino: {ruta_copia}")
    if os.path.isfile(ruta_copia):
        with open(ruta_copia, "rb") as copia:
            coincide = copia.read() == contenido_original
        print(f"¿El contenido coincide con el original? {coincide}")
    else:
        print("La ruta de destino no corresponde a un archivo.")
````

````python
# Solución alternativa con pathlib
from pathlib import Path

directorio_actual = Path.cwd()
ruta_original = directorio_actual / "mi_archivo.txt"

print("1. Archivo de ejemplo")
try:
    with ruta_original.open("x", encoding="utf-8") as archivo:
        archivo.write("Este es un archivo de ejemplo.\n")
except FileExistsError:
    print("El destino ya existe y no se sobrescribe.")

if not ruta_original.is_file():
    raise ValueError(f"La ruta original no corresponde a un archivo: {ruta_original}")

print(f"Archivo original: {ruta_original}")
print("Elementos del directorio de trabajo")
for elemento in sorted(directorio_actual.iterdir()):
    print(f"- {elemento.name}")

contenido_original = ruta_original.read_bytes()

# parent obtiene el directorio superior sin cambiar el de trabajo.
subcarpeta = directorio_actual / "nueva_carpeta"
carpeta_superior = directorio_actual.parent / "copias_superior"

destinos = [
    ("2. Copia en una subcarpeta", subcarpeta),
    ("3. Copia en una carpeta del directorio superior", carpeta_superior),
]

for descripcion, carpeta in destinos:
    print(f"\n{descripcion}")
    carpeta.mkdir(parents=True, exist_ok=True)
    ruta_copia = carpeta / ruta_original.name

    # "xb" conserva cualquier destino existente, a diferencia de write_bytes().
    try:
        with ruta_copia.open("xb") as copia:
            copia.write(contenido_original)
    except FileExistsError:
        print("El destino ya existe y se conserva sin cambios.")

    print(f"Ruta de destino: {ruta_copia}")
    if ruta_copia.is_file():
        coincide = ruta_copia.read_bytes() == contenido_original
        print(f"¿El contenido coincide con el original? {coincide}")
    else:
        print("La ruta de destino no corresponde a un archivo.")
````
