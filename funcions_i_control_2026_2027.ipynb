### Funciones definidas por el usuario

Crear nuestras propias funciones permite separar un problema en tareas más pequeñas y reutilizar cada solución con distintos datos. Por ejemplo, una función puede calcular la densidad de una muestra y utilizarse después con otras masas y volúmenes.

Una definición comienza con `def`, seguida del nombre de la función, sus **parámetros** entre paréntesis y dos puntos. El bloque indentado contiene las instrucciones que se ejecutarán cuando la llamemos. La sentencia `return` termina la llamada y entrega el resultado al código que la inició.

En este primer ejemplo, `valor` es el parámetro de la función. Al escribir `duplicar(3)`, el número `3` es el **argumento**, es decir, el dato concreto que proporcionamos en esa llamada.

````python
def duplicar(valor):
    """Devuelve el doble del valor numérico recibido."""
    resultado = valor * 2
    return resultado


# La definición prepara la función; esta llamada ejecuta sus instrucciones.
duplicar(3)
````

El resultado de `duplicar(3)` es `6`. El nombre `resultado` se utiliza dentro de la función y es una **variable local**. Para aprovechar el valor devuelto fuera de ella, podemos guardarlo con una asignación como `doble = duplicar(3)`.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #59A14F; background-color: rgba(89,161,79,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;"><code>return</code> y <code>print()</code> cumplen funciones distintas. El primero devuelve un valor que podemos reutilizar; el segundo lo muestra como texto. Si una función llega al final sin ejecutar un <code>return</code> con un valor, devuelve <code>None</code>.</p>
</div>

Para que una función sea fácil de entender y utilizar, conviene cuidar cuatro aspectos.

* **Nombre descriptivo**. Debe indicar la tarea que realiza o el resultado que obtiene, como `calcular_densidad`.
* **Parámetros claros**. Sus nombres y su documentación deben permitir reconocer los datos esperados y sus unidades.
* **Documentación útil**. La cadena de documentación o *docstring*, situada al comienzo del cuerpo, explica el propósito, los parámetros, el resultado y las restricciones relevantes. No es lo mismo que un comentario iniciado con `#`.
* **Comprobaciones pertinentes**. Cuando una entrada no es válida, podemos informar del problema mediante una excepción. Más adelante veremos cómo generarlas y manejarlas.

La siguiente función calcula la densidad mediante

$$
\rho = \frac{m}{V},
$$

donde $m$ es la masa en kilogramos, $V$ el volumen en metros cúbicos y $\rho$ la densidad en $\mathrm{kg\,m^{-3}}$. En este ejemplo admitimos masas no negativas y exigimos un volumen mayor que cero.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 3px solid #72A0C1; background-color: rgba(76,120,168,0.025); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Las anotaciones como <code>masa: float</code> indican el tipo de dato esperado, y <code>-&gt; float</code> describe el resultado. Ayudan a documentar el código y a revisarlo con herramientas, pero Python no comprueba ni convierte automáticamente los argumentos a esos tipos.</p>
</div>

````python
def calcular_densidad(masa: float, volumen: float) -> float:
    """Calcula la densidad a partir de la masa y el volumen.

    Parámetros
    ----------
    masa : float
        Masa no negativa, en kg.
    volumen : float
        Volumen estrictamente positivo, en m³.

    Devuelve
    --------
    float
        Densidad en kg/m³.

    Excepciones
    -----------
    ValueError
        Si la masa es negativa o el volumen no es positivo.
    """
    # Comprobamos las restricciones antes de realizar la división.
    if masa < 0:
        raise ValueError("La masa no puede ser negativa.")
    if volumen <= 0:
        raise ValueError("El volumen debe ser mayor que cero.")

    densidad = masa / volumen
    return densidad
````

````python
# Una masa de 3 kg y un volumen de 5 m³ dan una densidad de 0.6 kg/m³.
calcular_densidad(3, 5)
````

Una función también puede devolver varios datos agrupados. La siguiente variante devuelve la densidad junto con la masa y el volumen utilizados, en ese orden. La llamamos `calcular_densidad_con_datos` para conservar la función anterior.

La instrucción `return densidad, masa, volumen` construye una **tupla**. Podemos guardarla completa o **desempaquetarla**, asignando sus tres elementos a tres nombres. La nueva función reutiliza `calcular_densidad()`, incluidas sus comprobaciones, sin repetir el cálculo.

````python
def calcular_densidad_con_datos(
    masa: float, volumen: float
) -> tuple[float, float, float]:
    """Devuelve la densidad y los datos utilizados para calcularla.

    Recibe la masa en kg y el volumen en m³. El resultado es una tupla
    con la densidad en kg/m³, la masa y el volumen, en ese orden.
    Se aplican las mismas restricciones que en calcular_densidad().
    """
    densidad = calcular_densidad(masa, volumen)
    return densidad, masa, volumen
````

````python
resultado = calcular_densidad_con_datos(3, 5)
print("Tupla devuelta", resultado)

# El orden de los nombres coincide con el de los elementos de la tupla.
densidad, masa, volumen = resultado
print(f"Densidad = {densidad} kg/m³")
print(f"Masa = {masa} kg")
print(f"Volumen = {volumen} m³")
````

## Estructuras de control condicionales

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Un <strong>condicional</strong> permite elegir qué instrucciones se ejecutan según se cumpla o no una condición. Por ejemplo, podemos mostrar un mensaje distinto dependiendo de la temperatura.</p>
</div>

La instrucción `if` evalúa una condición. Si es verdadera, ejecuta el bloque indentado que aparece debajo; si es falsa, lo omite.

En el siguiente ejemplo, la condición `temperatura > 25` comprueba si la temperatura supera los 25 °C. Los umbrales de esta sección son ilustrativos y no representan una clasificación universal del confort térmico.

````python
temperatura = 35  # °C

if temperatura > 25:
    print("La temperatura supera los 25 °C.")
````

La instrucción `elif`, abreviatura de *else if*, introduce otra condición que solo se evalúa si las anteriores han sido falsas. Podemos incluir varios bloques `elif`, pero en una misma cadena se ejecuta únicamente el primero cuya condición se cumpla.

La comparación encadenada `15 <= temperatura <= 25` comprueba que la temperatura está entre 15 y 25 °C, incluidos ambos extremos.

````python
temperatura = 21  # °C

if temperatura > 25:
    print("La temperatura supera los 25 °C.")
elif 15 <= temperatura <= 25:
    print("La temperatura está entre 15 y 25 °C, incluidos ambos extremos.")
````

El bloque `else` recoge los casos que no han cumplido ninguna de las condiciones anteriores. No lleva una condición propia y es opcional. En este ejemplo, corresponde a temperaturas inferiores a 15 °C.

Si usamos varios `if` independientes en lugar de una cadena `if`–`elif`–`else`, cada condición se evalúa por separado y podrían ejecutarse varios bloques.

````python
temperatura = 10  # °C

if temperatura > 25:
    print("La temperatura supera los 25 °C.")
elif 15 <= temperatura <= 25:
    print("La temperatura está entre 15 y 25 °C, incluidos ambos extremos.")
else:
    print("La temperatura es inferior a 15 °C.")
````

Python evalúa `and` y `or` de izquierda a derecha y se detiene cuando ya puede determinar el resultado. Este comportamiento se llama **evaluación en cortocircuito**. Por ejemplo, en `divisor != 0 and 10 / divisor > 2`, no se calcula la división si `divisor` es cero.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 3px solid #72A0C1; background-color: rgba(76,120,168,0.025); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;"><code>and</code> y <code>or</code> también admiten otros objetos y devuelven uno de sus operandos, no necesariamente un <code>bool</code>. Por ejemplo, <code>0 or 5</code> devuelve <code>5</code>. En los ejemplos de esta sección combinamos comparaciones, por lo que sus resultados sí son booleanos.</p>
</div>

#### Operadores lógicos y operadores a nivel de bit

Los operadores `&`, `|`, `^` y `~` actúan sobre los bits de los enteros. No deben confundirse con `and`, `or` y `not`, ni con `!=`, que es un operador de comparación.

Cuando combinamos valores `True` y `False`, `&` y `|` pueden dar el mismo resultado que `and` y `or`, pero **no realizan evaluación en cortocircuito**. Para combinar condiciones individuales, utilizaremos normalmente `and`, `or` y `not`.

Las bibliotecas de análisis de datos pueden dar a estos operadores un significado elemento a elemento. En ese contexto es importante encerrar cada comparación entre paréntesis. La [referencia de tipos de Python](https://docs.python.org/3/library/stdtypes.html#boolean-operations-and-or-not) detalla el comportamiento de los operadores del lenguaje.

En las dos celdas siguientes aplicamos las mismas tres condiciones a una temperatura, una humedad relativa y una velocidad del viento. La primera utiliza `and`; la segunda muestra el contraste con `&`.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #C08B3E; background-color: rgba(192,139,62,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Los umbrales de 30 °C, 30 % y 30 km/h sirven aquí para practicar una condición compuesta relacionada con el tiempo favorable a incendios. No constituyen por sí solos una predicción de incendio ni una evaluación completa del peligro.</p>

## Bucles `for` y `while`

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Un <strong>bucle</strong> repite un bloque de instrucciones. Cada repetición se llama <strong>iteración</strong>. Los bucles permiten, por ejemplo, aplicar una comprobación a cada temperatura de una serie sin escribir la misma instrucción para cada dato.</p>
</div>

Python dispone de dos estructuras de bucle, `while` y `for`. La primera repite una operación mientras se cumpla una condición; la segunda recorre los elementos de un objeto iterable.

### Bucle `while`

La condición de un `while` se comprueba **antes de cada iteración**. Si es falsa desde el principio, el bloque no se ejecuta ninguna vez.

En el siguiente ejemplo, `contador` comienza en cero y aumenta una unidad en cada paso. Se muestran los valores de `0` a `4`; cuando llega a `5`, la condición deja de cumplirse y el bucle termina.

````python
contador = 0

while contador < 5:
    print(contador)
    contador += 1  # Equivale a contador = contador + 1.
````

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #C08B3E; background-color: rgba(192,139,62,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Un <code>while</code> necesita una forma de terminar, por ejemplo, que su condición llegue a ser falsa o que se ejecute un <code>break</code>. En el ejemplo anterior, si elimináramos el incremento de <code>contador</code>, la condición seguiría siendo verdadera y el bucle no acabaría por sí solo.</p>
</div>

### Bucle `for`

Un `for` recorre un **iterable**, es decir, un objeto del que podemos obtener elementos uno a uno. Las listas, las tuplas y las cadenas son ejemplos de secuencias iterables. Los diccionarios y los conjuntos también son iterables, aunque no son secuencias.

En este ejemplo, `temperatura` toma sucesivamente cada valor de la lista. No necesitamos crear ni actualizar un contador.

```python
temperaturas = [18, 20, 22]  # °C

for temperatura in temperaturas:
    print(f"{temperatura} °C")
```

Al recorrer un diccionario obtenemos sus claves de forma predeterminada. En un conjunto no debemos depender de un orden concreto de recorrido. Para repetir una operación un número determinado de veces, podemos usar `range()`; por ejemplo, `for i in range(5)` recorre los enteros de `0` a `4`.

### Control del flujo dentro de los bucles

Podemos modificar el recorrido de un bucle con dos instrucciones.

* **`break`** termina el bucle en el que se encuentra. Si hay bucles anidados, sale únicamente del más interno que lo contiene.
* **`continue`** omite el resto de la iteración actual y pasa a la siguiente. En un `while`, esto implica volver a comprobar la condición.

El primer ejemplo utiliza `continue` para omitir temperaturas inferiores a 20 °C. Los valores que alcanzan ese umbral llegan a la instrucción `print()`.

El segundo busca una variable por su nombre. `enumerate()` proporciona en cada paso su posición, empezando por cero, y su valor. Cuando encuentra una coincidencia, `break` termina la búsqueda.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 3px solid #72A0C1; background-color: rgba(76,120,168,0.025); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Un bucle también puede llevar un <code>else</code>. En un <code>for</code>, ese bloque se ejecuta al agotar el recorrido sin haber salido con <code>break</code>. En el ejemplo, permite informar de que no se ha encontrado la variable. El <code>else</code> está alineado con <code>for</code>, no con el <code>if</code> interior. Si una excepción o un <code>return</code> interrumpen el bucle, tampoco se alcanza ese <code>else</code>.</p>
</div>

````python
temperaturas = [18, 20, 22, 17, 24, 21]  # °C

for temperatura in temperaturas:
    if temperatura < 20:
        continue
    print(f"Temperatura de al menos 20 °C — {temperatura} °C")
````

````python
variables = ["prec", "time", "temp", "wind"]
variable_buscada = "temp"

for posicion, nombre in enumerate(variables):
    if nombre == variable_buscada:
        print(f"'{variable_buscada}' encontrada en la posición {posicion}.")
        break
else:
    # Este bloque se alcanza si se agota la lista sin encontrar coincidencias.
    print(f"'{variable_buscada}' no se encuentra en la lista.")
````

## Introducción al manejo de errores en Python

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Una <strong>excepción</strong> señala que una operación no puede completarse de la forma prevista. Por ejemplo, puede aparecer al dividir entre cero o al intentar acceder a un dato que no existe.</p>
</div>

Python genera muchas excepciones automáticamente. También podemos provocarlas con `raise` para indicar que una entrada incumple las condiciones de nuestra función, como hicimos al comprobar la masa y el volumen.

Una excepción interrumpe el flujo normal y busca un bloque que pueda manejarla. Si no lo encuentra, la llamada termina con un mensaje de error. Las instrucciones `try`, `except`, `else` y `finally` permiten organizar la respuesta a estas situaciones.

### Uso de `raise`

La sentencia `raise` lanza una excepción de forma explícita. El tipo indica la clase de problema, mientras que el mensaje aporta información sobre lo ocurrido.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #C08B3E; background-color: rgba(192,139,62,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">La siguiente celda produce un <code>TypeError</code> de forma intencionada. La última línea de la salida muestra el tipo de excepción y el mensaje; las líneas anteriores ayudan a localizar la instrucción que la ha provocado.</p>
</div>

````python
# Error intencionado para observar el tipo de excepción y su mensaje.
raise TypeError("El argumento debe ser numérico.")
````

En este caso, `TypeError` es la clase de excepción y `"El argumento debe ser numérico."` es el mensaje que hemos elegido.

También podemos lanzar una excepción únicamente cuando se incumpla una condición. En las dos celdas siguientes asignamos un valor negativo a `x` y comprobamos si es válido para una operación que requiere valores no negativos. El `ValueError` resultante también es intencionado.

````python
x = -3
````

````python
# Error intencionado porque el valor de x es negativo.
if x < 0:
    raise ValueError("x no puede ser negativo.")
````

### Métodos para el control de errores

**`try` y `except`** permiten intentar una operación y responder a un tipo concreto de excepción. Las instrucciones que pueden fallar se colocan en `try`, y la respuesta se define en el bloque `except` correspondiente.

Cuando aparece la excepción, se omite el resto del bloque `try`. Si un `except` coincide con su tipo, ese bloque la maneja. Es preferible indicar qué excepción esperamos en lugar de capturar cualquier error y ocultar posibles fallos del programa.

````python
try:
    resultado = 10 / 0
except ZeroDivisionError:
    # Manejamos este problema concreto con un mensaje explicativo.
    print("No se puede dividir entre cero.")
````

**`finally`** contiene instrucciones que deben ejecutarse al abandonar el bloque `try`, tanto si la operación tiene éxito como si aparece una excepción. Se utiliza, por ejemplo, para liberar un recurso que ya no necesitamos.

El bloque `finally` no captura ni elimina por sí mismo una excepción. Si no hay un `except` que la maneje, el error se propaga después de ejecutar `finally`. En este ejemplo, el mensaje final aparece aunque la división haya fallado.

````python
try:
    resultado = 10 / 0
except ZeroDivisionError:
    print("No se puede dividir entre cero.")
finally:
    print("Se ha terminado el intento de división.")
````

**`else`**, cuando acompaña a `try` y `except`, se ejecuta si el bloque `try` finaliza con normalidad, sin lanzar una excepción. Permite separar la operación que puede fallar de las instrucciones que solo tienen sentido cuando ha funcionado.

En esta ocasión el divisor es distinto de cero, por lo que la división se completa y se muestra su resultado.

````python
try:
    resultado = 10 / 2
except ZeroDivisionError:
    print("No se puede dividir entre cero.")
else:
    print(f"La división se ha completado. Resultado = {resultado}")
````

## El coste computacional de las funciones

El **coste computacional** describe los recursos que necesita un cálculo, como el tiempo de ejecución o la memoria. Aquí nos centraremos en el tiempo, especialmente relevante cuando procesamos muchas observaciones o repetimos una función sobre series meteorológicas extensas.

Una pequeña diferencia por llamada puede acumularse. Si una función tarda un segundo y la ejecutamos mil veces de forma secuencial, necesitaremos unos 16 minutos y 40 segundos, sin contar otros costes del programa.

### Medir tiempos con `%timeit`

La función mágica `%timeit` de IPython ejecuta una instrucción repetidamente para estimar cuánto tarda. El resultado habitual muestra la **media y la desviación estándar de los tiempos por ejecución obtenidos en varias repeticiones**. No muestra únicamente el mejor tiempo ni una media de las ejecuciones más rápidas.

Podemos controlar la medición mediante dos opciones.

* **`-n`** fija cuántas ejecuciones se realizan dentro de cada repetición.
* **`-r`** fija cuántas veces se repite ese grupo de ejecuciones.

Por ejemplo, `%timeit -n 1000 -r 5 calcular_densidad(3, 5)` realiza cinco repeticiones de mil llamadas cada una. Si no indicamos `-n`, IPython elige automáticamente un número de ejecuciones adecuado. La [documentación de `%timeit`](https://ipython.readthedocs.io/en/stable/interactive/magics.html#magic-timeit) explica las demás opciones.

En el siguiente ejemplo comparamos una suma directa con dos funciones que calculan el mismo resultado. La segunda crea una lista y la recorre, de modo que realiza operaciones adicionales para una tarea muy sencilla.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 3px solid #72A0C1; background-color: rgba(76,120,168,0.025); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Usamos variables en la suma directa para que la comparación mida una operación con sus valores. Una expresión formada solo por constantes, como <code>1 + 1</code>, puede resolverse durante la compilación y no medir la suma que pretendíamos comparar. Los tiempos observados también dependen del ordenador, de la versión de Python y de otras tareas en ejecución.</p>
</div>

````python
def suma_dos_num(x, y):
    """Devuelve la suma de dos valores numéricos."""
    return x + y


def suma_dos_num_con_bucle(x, y):
    """Obtiene la misma suma mediante una lista y un bucle."""
    numeros = [x, y]
    total = 0
    for numero in numeros:
        total += numero
    return total


sumando_1 = 1
sumando_2 = 1

print("Suma directa")
%timeit -n 100000 -r 5 sumando_1 + sumando_2

print("Suma mediante una función")
%timeit -n 100000 -r 5 suma_dos_num(sumando_1, sumando_2)

print("Suma mediante una función con lista y bucle")
%timeit -n 100000 -r 5 suma_dos_num_con_bucle(sumando_1, sumando_2)
````

### Breve apunte sobre benchmarking en Python

El **benchmarking** consiste en medir el rendimiento de una tarea bajo condiciones concretas, por ejemplo, para comparar dos implementaciones con los mismos datos. Antes de interpretar sus tiempos, debemos comprobar que ambas producen resultados equivalentes.

El **perfilado** o *profiling* responde a una pregunta diferente. Permite saber cómo se reparte el tiempo dentro de un programa y localizar las funciones que concentran una parte importante del trabajo. Esas zonas pueden convertirse en **cuellos de botella**, porque limitan el rendimiento del conjunto.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #59A14F; background-color: rgba(89,161,79,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Conviene comenzar con una solución clara y correcta, medirla y optimizar las partes que realmente lo necesitan. Una versión más breve no es necesariamente más rápida, y una mejora pequeña no siempre compensa una pérdida de legibilidad.</p>
</div>

````python
import cProfile


def sumar_cuadrados(limite: int) -> int:
    """Suma los cuadrados de los enteros desde 0 hasta limite - 1."""
    total = 0
    for numero in range(limite):
        total += numero ** 2
    return total


cProfile.run("sumar_cuadrados(100_000)", sort="cumulative")
````

Al trabajar en Jupyter también pueden aparecer llamadas internas del entorno. Para interpretar el ejemplo, podemos centrarnos en las filas de las funciones que hemos definido.

Los tiempos se expresan en segundos. Las dos columnas `percall` dividen los tiempos correspondientes entre el número de llamadas pertinente. Una fila con `0.000` puede tener un tiempo pequeño que queda oculto por el redondeo, no un coste exactamente nulo.

El perfilado añade trabajo de medición, por lo que sus tiempos no equivalen a los de una ejecución sin instrumentar. Para comparar duraciones globales, podemos complementar el [informe de `cProfile`](https://docs.python.org/3/library/profile.html) con `%timeit`.

A continuación utilizaremos un ejemplo con llamadas entre funciones. Primero definimos `es_primo()`, que comprueba si un entero es primo. La función `math.isqrt(n)` obtiene exactamente la parte entera de la raíz cuadrada de un entero no negativo, sin pasar por una aproximación de coma flotante.

````python
from math import isqrt


def es_primo(n: int) -> bool:
    """Indica si el entero n es primo; los valores menores que 2 no lo son."""
    if n < 2:
        return False

    # range excluye su límite superior; +1 incluye la raíz si es entera.
    for divisor in range(2, isqrt(n) + 1):
        if n % divisor == 0:
            return False

    # Solo llegamos aquí si no hemos encontrado ningún divisor.
    return True
````

### Por qué basta con comprobar hasta la raíz cuadrada

Si un entero $n > 1$ no es primo, puede escribirse como $n = a\,b$, con dos divisores enteros mayores que uno. Al menos uno de ellos debe ser menor o igual que $\sqrt{n}$.

Si ambos fueran mayores que $\sqrt{n}$, su producto sería mayor que $n$, lo que contradice $n = a\,b$. Por tanto, no necesitamos probar todos los números hasta $n-1$ para encontrar un divisor.

Por ejemplo, $49 = 7 \times 7$. Debemos incluir el $7$ en la comprobación, de ahí el `+1` en `range(2, isqrt(n) + 1)`. El operador `%` calcula el resto de la división; si `n % divisor == 0`, hemos encontrado un divisor exacto y la función devuelve `False` inmediatamente.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #59A14F; background-color: rgba(89,161,79,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">En el peor caso comprobamos aproximadamente <code>√n</code> candidatos en lugar de <code>n</code>. Esta reducción del número de comprobaciones es una mejora del algoritmo, no solo una forma más breve de escribirlo.</p>
</div>

Una vez disponible `es_primo()`, podemos reutilizarla para construir una lista con todos los primos menores que un límite dado.

````python
def primos_hasta(limite: int) -> list[int]:
    """Devuelve los primos menores que el límite entero, sin incluirlo.

    Si el límite es menor o igual que 2, devuelve una lista vacía.
    """
    return [numero for numero in range(2, limite) if es_primo(numero)]
````

La función anterior utiliza una **comprensión de lista**, una forma compacta de construir una lista recorriendo elementos y, opcionalmente, seleccionando algunos de ellos. Su estructura general es la siguiente.

```text
[expresion for elemento in iterable if condicion]
```

En cada paso, `elemento` toma un valor del iterable. Si se cumple la condición, se evalúa `expresion` y su resultado se añade a la lista. La parte `if condicion` puede omitirse cuando queremos incluir todos los elementos.

En `primos_hasta()`, recorremos los enteros desde `2` hasta `limite - 1` y conservamos únicamente los que cumplen `es_primo(numero)`. La misma operación puede escribirse de forma más explícita con un bucle.

```python
limite = 10
primos = []

for numero in range(2, limite):
    if es_primo(numero):
        primos.append(numero)

print(primos)  # [2, 3, 5, 7]
```

Ambas formas siguen el mismo procedimiento. La comprensión evita algunas líneas, pero no cambia el algoritmo de búsqueda. En la siguiente celda perfilamos `primos_hasta(10000)` para observar también sus llamadas a `es_primo()`.

````python
import cProfile

cProfile.run("primos_hasta(10000)", sort="cumulative")
````

Para medir la duración global de la función, podemos utilizar `%timeit` o un reloj del módulo `time`. La primera opción repite el cálculo; la segunda permite medir un bloque una sola vez y funciona también en un programa de Python fuera de Jupyter.

En las próximas celdas obtenemos primero los primos menores que `1000`, repetimos la medición con `%timeit` y después calculamos el tiempo transcurrido con `time.perf_counter()`. Este reloj está pensado para medir duraciones; debemos restar dos lecturas, ya que su valor aislado no representa una fecha ni una hora del día.

La impresión de la lista queda fuera de las mediciones para no incluir el tiempo de mostrar los resultados. Una medición aislada es orientativa y puede variar entre ejecuciones.

````python
primos = primos_hasta(1000)
print(primos)
````

````python
%timeit -n 100 -r 5 primos_hasta(1000)
````

````python
import time

inicio = time.perf_counter()
primos = primos_hasta(1000)
fin = time.perf_counter()

# La diferencia entre las dos lecturas es la duración en segundos.
tiempo_transcurrido = fin - inicio
print(f"La función tardó {tiempo_transcurrido:.6f} segundos en ejecutarse.")
````

### 1. Precipitación acumulada

Escribe una función que reciba una lista con la precipitación diaria de **14 días**, expresada en milímetros, y devuelva la precipitación acumulada durante ese periodo, también en milímetros.

Puedes utilizar la siguiente lista para comprobar tu función. El resultado esperado es `37.0` mm.

```python
precipitacion_diaria = [
    0.0, 3.2, 0.0, 5.6, 12.4, 0.0, 0.0,
    1.8, 7.0, 0.0, 2.5, 0.0, 0.0, 4.5,
]
```

````python
def precipitacion_acum(precipitacion):
    """ Suma los valores de precipitación diaria"""
    resultado = 0
    for i in precipitacion:
        resultado = resultado + i
    return resultado

precipitacion_diaria = [
    0.0, 3.2, 0.0, 5.6, 12.4, 0.0, 0.0,
    1.8, 7.0, 0.0, 2.5, 0.0, 0.0, 4.5,
]

precipitacion_acum(precipitacion_diaria)
````

### 4. Letras mayúsculas y minúsculas

Escribe una función que reciba una frase y devuelva cuántas letras mayúsculas y cuántas minúsculas contiene. Los espacios, las cifras y los signos de puntuación no deben sumarse a ninguno de los dos recuentos.

Por ejemplo, `"Hola, Meteo 2026!"` contiene **2 mayúsculas y 7 minúsculas**. La función debe devolver ambos recuentos para que puedan utilizarse fuera de ella.

````python
def contador_maj_min(frase: str):
    mayusculas = 0
    minusculas = 0
    
    for i in frase:
        if i.isupper():
            mayusculas += 1
        elif i.islower():
            minusculas += 1
            
    return {"mayúsculas": mayusculas, "minúsculas": minusculas}

frase_1 = "Hola, Mateo 2026!"
contador_maj_min(frase_1)
````

### 5. Conversión de temperaturas

Escribe una función que convierta temperaturas entre las escalas Celsius y Fahrenheit. La entrada será una cadena como `"45F"` o `"32C"`, y el resultado debe indicar la temperatura en la otra escala junto con su unidad.

Las relaciones entre los valores numéricos de ambas escalas son

$$
T_{\mathrm{F}} = \frac{9}{5}T_{\mathrm{C}} + 32,
\qquad
T_{\mathrm{C}} = \frac{5}{9}\left(T_{\mathrm{F}} - 32\right),
$$

donde $T_{\mathrm{C}}$ es la temperatura expresada en °C y $T_{\mathrm{F}}$ la temperatura expresada en °F. Comprueba que `"0C"` corresponde a 32 °F y `"32F"` a 0 °C. La parte numérica de la entrada también puede contener decimales o un signo negativo.

````python
def cambio_escala_temperatura(Temperatura: str):
    valor_temp = int(Temperatura[:-1])
    if Temperatura[-1] == "C":
        temp_F = ((9/5)*valor_temp)+32
        temp_F_str = str(temp_F)+"F"
        return temp_F_str
    if Temperatura[-1] == "F":
        temp_C = (5/9)*(valor_temp-32)
        temp_C_str = str(temp_C)+"C"
        return temp_C_str
    else:
        print("Valor no válido")
        return None

print(cambio_escala_temperatura("0C"))
print(cambio_escala_temperatura("32F"))
````
