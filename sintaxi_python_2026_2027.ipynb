## ¿Qué es Python?

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Python es un <strong>lenguaje de programación</strong>: una forma de escribir instrucciones que un ordenador puede ejecutar. Es de <strong>alto nivel</strong>, porque permite expresar las operaciones sin gestionar directamente muchos detalles de la máquina, y de <strong>propósito general</strong>, porque sirve para tareas muy diversas.</p>
</div>

Suele describirse como un lenguaje **interpretado**, porque trabajamos con un intérprete, el programa encargado de ejecutar nuestro código. Esto permite probar instrucciones de forma interactiva, como haremos en JupyterLab, sin preparar manualmente un ejecutable para cada ejemplo.

   Python favorece una sintaxis legible, con pocos símbolos para delimitar las instrucciones. Esta característica facilita la lectura, la corrección y el mantenimiento del código, aunque escribir programas claros también depende de cómo los organicemos. Si venís de lenguajes como C++ o Java, reconoceréis muchas ideas, pero encontraréis diferencias en la forma de expresarlas.

   En ciencia e ingeniería se utiliza para procesar y analizar datos, realizar cálculos numéricos y generar gráficos. Una parte importante de estas posibilidades procede de sus **bibliotecas**, conjuntos de herramientas que podemos incorporar a nuestros programas. En meteorología y climatología, por ejemplo, nos permiten trabajar con series de temperatura, presión, humedad o velocidad del viento.

## Estructura de un programa de Python

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Un programa de Python está formado por <strong>instrucciones</strong> que indican al ordenador qué debe hacer. En programación, estas instrucciones reciben habitualmente el nombre de <strong>sentencias</strong>.</p>
</div>

Por defecto, Python ejecuta las sentencias en el orden en que aparecen, de arriba abajo. Sin embargo, algunas estructuras permiten modificar este flujo, por ejemplo, para **tomar decisiones**, **repetir operaciones** o **agrupar instrucciones** que podremos utilizar cuando las necesitemos.

   Habitualmente escribimos una sentencia por línea, aunque una misma sentencia puede ocupar varias líneas cuando es larga o cuando dividirla mejora la legibilidad del código.

   Para entender mejor cómo se construye un programa, conviene distinguir entre **sentencias** y **expresiones**:

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 3px solid #72A0C1; background-color: rgba(76,120,168,0.025); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0 0 0.4em 0; font-weight: 600;">Sentencia de asignación</p><p style="margin: 0.45em 0;">Asocia un nombre con un valor para poder utilizarlo posteriormente, como en este ejemplo.</p>
<pre style="margin: 0.65em 0; padding: 0.7em 0.85em; background-color: rgba(127,127,127,0.09); border-radius: 4px; overflow-x: auto;"><code class="language-python" style="background-color: transparent; padding: 0; color: inherit;">x = 10
</code></pre>
<p style="margin: 0.45em 0;">Esta sentencia hace que el nombre <code>x</code> quede asociado al valor <code>10</code>. La asignación no equivale a declarar previamente el tipo de una variable, como ocurre en algunos otros lenguajes.</p>
</div>

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 3px solid #72A0C1; background-color: rgba(76,120,168,0.025); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0 0 0.4em 0; font-weight: 600;">Expresión</p><p style="margin: 0.45em 0;">Es una parte del código que Python puede evaluar para obtener un valor. Puede ser algo tan sencillo como un número.</p>
<pre style="margin: 0.65em 0; padding: 0.7em 0.85em; background-color: rgba(127,127,127,0.09); border-radius: 4px; overflow-x: auto;"><code class="language-python" style="background-color: transparent; padding: 0; color: inherit;">10
</code></pre>
<p style="margin: 0.45em 0;">También puede incluir variables, operadores y llamadas a funciones, como en el siguiente ejemplo.</p>
<pre style="margin: 0.65em 0; padding: 0.7em 0.85em; background-color: rgba(127,127,127,0.09); border-radius: 4px; overflow-x: auto;"><code class="language-python" style="background-color: transparent; padding: 0; color: inherit;">x * 2
</code></pre>
<p style="margin: 0.45em 0;">Si <code>x</code> vale <code>10</code>, esta expresión produce el valor <code>20</code>. Las expresiones también pueden formar parte de una sentencia, como en esta asignación.</p>
<pre style="margin: 0.65em 0; padding: 0.7em 0.85em; background-color: rgba(127,127,127,0.09); border-radius: 4px; overflow-x: auto;"><code class="language-python" style="background-color: transparent; padding: 0; color: inherit;">y = x * 2
</code></pre>
<p style="margin: 0.45em 0;">Aquí, Python evalúa primero <code>x * 2</code> y después asocia el resultado al nombre <code>y</code>.</p>
</div>

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 3px solid #72A0C1; background-color: rgba(76,120,168,0.025); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0 0 0.4em 0; font-weight: 600;">Sentencia compuesta</p><p style="margin: 0.45em 0;">Reúne otras sentencias dentro de uno o varios bloques de código. Entre las más habituales se encuentran los condicionales (<code>if</code>, con posibles ramas <code>elif</code> y <code>else</code>), los bucles (<code>for</code>, <code>while</code>) y las definiciones de funciones (<code>def</code>).</p>
<p style="margin: 0.45em 0;">Un condicional permite ejecutar una parte del código únicamente cuando se cumple una determinada condición.</p>
<pre style="margin: 0.65em 0; padding: 0.7em 0.85em; background-color: rgba(127,127,127,0.09); border-radius: 4px; overflow-x: auto;"><code class="language-python" style="background-color: transparent; padding: 0; color: inherit;">if x &gt; 0:
    print("x es positivo")
</code></pre>
<p style="margin: 0.45em 0;">En este ejemplo, <code>print()</code> muestra el mensaje entre comillas solo si <code>x</code> es mayor que cero.</p>
</div>

Una expresión también puede escribirse por sí sola. En ese caso hablamos de una **sentencia de expresión**. Esto resulta especialmente útil cuando trabajamos en entornos interactivos como Jupyter.

   Por ejemplo, podemos escribir esta expresión al final de una celda de código.

   ```python
   3 + 2
   ```

   Jupyter evalúa la expresión y muestra su resultado.

   ```text
   5
   ```

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #59A14F; background-color: rgba(89,161,79,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Esta forma de trabajo facilita la exploración paso a paso, ya que podemos realizar una operación, observar el resultado y continuar a partir de él sin utilizar <code>print()</code> en cada comprobación. Por defecto, se muestra el valor de la expresión situada al final de la celda, siempre que no sea <code>None</code>, el valor que Python utiliza para representar la ausencia de un resultado.</p>
</div>

## Importancia de la indentación en Python

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Un <strong>bloque de código</strong> es un conjunto de instrucciones que pertenecen a una misma estructura, por ejemplo, las que se ejecutan dentro de un condicional, un bucle o una función. En Python, estos bloques se identifican mediante la <strong>indentación</strong>, es decir, el espacio que dejamos al comienzo de las líneas.</p>
</div>

En muchos lenguajes de programación se utilizan llaves `{}` u otros símbolos para delimitar los bloques. Python emplea la propia indentación, de modo que la disposición visual del código no es solo una cuestión de estilo, sino que refleja su **estructura lógica**.

   Las instrucciones de un mismo bloque deben quedar alineadas. Cuando una estructura contiene otra, añadimos un nivel de indentación para distinguir el bloque interior. Por ejemplo, un bucle dentro de una función estará más indentado que la cabecera de esa función.

   Por convención, y de acuerdo con la guía de estilo **PEP 8**, se utilizan **4 espacios por nivel de indentación**. Python también admite tabulaciones, pero se recomienda trabajar con espacios y no mezclar ambos sistemas.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #59A14F; background-color: rgba(89,161,79,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0 0 0.4em 0; font-weight: 600;">Idea clave</p><p style="margin: 0.45em 0;">Cambiar la indentación puede producir un error o modificar qué instrucciones se ejecutan. Un código puede estar bien escrito desde el punto de vista sintáctico y, aun así, hacer algo distinto de lo que pretendíamos.</p>
</div>

Veámoslo con una función que calcula el promedio de una lista de temperaturas. Utilizaremos datos ficticios de tres ciudades, expresados en **°C**.

   Primero preparamos los datos, después definimos la función y, por último, la llamamos para cada ciudad. Las listas, los diccionarios y las funciones aparecerán de nuevo más adelante; por ahora, fijaos en **qué líneas están alineadas y cuáles pertenecen a un bloque interior**.

````python
# Datos ficticios: siete temperaturas diarias (°C) para cada ciudad.
registros_ciudades = [
    {"ciudad": "Madrid", "temperaturas": [22, 24, 26, 28, 23, 25, 24]},
    {"ciudad": "Barcelona", "temperaturas": [21, 23, 24, 25, 23, 22, 24]},
    {"ciudad": "Sevilla", "temperaturas": [30, 31, 32, 31, 30, 29, 31]},
]
````

````python
def calcular_promedio_temperaturas(lista_temperaturas):
    """Calcula el promedio de una lista numérica no vacía de temperaturas."""
    total = 0
    for temperatura in lista_temperaturas:
        total += temperatura  # Acumulamos una temperatura en cada vuelta.

    # Estas líneas se ejecutan una vez, después de terminar el bucle.
    promedio = total / len(lista_temperaturas)
    return promedio
````

La línea que empieza por `def` es la **cabecera** de la función. Termina en dos puntos, y las instrucciones de su cuerpo quedan un nivel más indentadas. Tanto la cabecera como la asignación de los datos anteriores comienzan al margen izquierdo, sin que ninguna esté dentro de la otra. La frase entre triples comillas es una cadena de documentación (*docstring*) que describe la función y no interviene en el cálculo.

   Dentro de la función, `total` empieza en cero. El bucle `for` recorre las temperaturas una a una y, en cada vuelta, `total += temperatura` añade el valor correspondiente a la suma. Esa línea está un nivel más indentada que `for` porque pertenece al bucle. Aquí, `+=` equivale a escribir `total = total + temperatura`.

   Las líneas que calculan y devuelven `promedio` están **fuera del bucle, pero dentro de la función**. Así, la división se realiza cuando ya se han sumado todas las temperaturas. `len(lista_temperaturas)` proporciona el número de datos y `return` devuelve el resultado a quien haya llamado a la función.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #C18A36; background-color: rgba(193,138,54,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Definir la función no calcula todavía ningún promedio. El cálculo se realiza cuando la <strong>llamamos</strong>. La lista debe contener al menos una temperatura; de lo contrario, su longitud sería cero y esta división no permitiría obtener un promedio.</p>
</div>

````python
for datos in registros_ciudades:
    # Cada vuelta trabaja con los datos de una ciudad.
    temperaturas_ciudad = datos["temperaturas"]
    promedio = calcular_promedio_temperaturas(temperaturas_ciudad)

    print(f"Promedio de {datos['ciudad']}: {promedio:.1f} °C")
````

Todas las instrucciones indentadas bajo este `for` se ejecutan para cada ciudad de `registros_ciudades`. La función recibe sus temperaturas, calcula el promedio y devuelve el valor que después mostramos.

   Si `print()` quedara fuera del bucle, el programa no daría necesariamente un error, pero solo mostraría el resultado de la última ciudad. Esta diferencia ayuda a entender por qué la indentación forma parte del significado del código.

   El texto precedido de `f` permite insertar valores entre llaves. Aquí usamos `:.1f` para mostrar el promedio con un decimal; veremos esta notación con más detalle en el apartado de formato de texto.

## Otras consideraciones

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Un código legible combina <strong>nombres claros</strong>, <strong>comentarios útiles</strong> y una <strong>presentación coherente</strong>. Estas decisiones ayudan tanto a quien lo escribe como a quien tenga que revisarlo más adelante.</p>
</div>

Los **comentarios** comienzan con `#`, fuera de una cadena de texto, y continúan hasta el final de la línea. Pueden ocupar una línea propia o aparecer después del código. Conviene usarlos para explicar el propósito de una operación, sus unidades o una decisión que no resulte evidente, sin repetir innecesariamente lo que ya se lee en la instrucción.

   Es preferible escribir **una sentencia por línea**. Python permite separar varias sentencias simples con `;`, como en `a = 3; b = 2; c = (a + b) / 2; print(c)`, pero esa presentación suele ser menos legible. En cambio, una sentencia larga puede repartirse entre varias líneas dentro de paréntesis, corchetes o llaves, como hemos hecho al preparar los datos.

   Para los **nombres de variables**, utilizad palabras que indiquen su propósito, por ejemplo, `temperatura_media`. Pueden contener letras, el carácter de subrayado `_` y números, pero **no pueden comenzar por un número**. Python 3 también admite muchas letras Unicode, como la `ñ` de `num_pequeño`.

   Se distinguen las mayúsculas de las minúsculas (*case sensitive*), de modo que `temperatura` y `Temperatura` son nombres diferentes.

   Las **palabras reservadas** tienen una función en el lenguaje y no pueden utilizarse como nombres de variables. Algunos ejemplos son `and`, `break`, `class`, `continue`, `def`, `del`, `elif`, `else`, `except`, `finally`, `for`, `from`, `global`, `if`, `import`, `in`, `is`, `lambda`, `not`, `or`, `pass`, `raise`, `return`, `try` y `while`. También están reservadas `True`, `False` y `None`; esta lista no es exhaustiva.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #C18A36; background-color: rgba(193,138,54,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">En Python 3, <code>print</code> y <code>exec</code> son funciones incorporadas, no palabras reservadas. Aun así, evitad emplearlas como nombres de variables, igual que <code>list</code>, <code>str</code>, <code>len</code> o <code>type</code>, ya que ese nombre dejaría de dar acceso a la función correspondiente mientras mantuviera la nueva asignación.</p>
</div>

   El término *booleano* procede de George Boole, cuyo trabajo sobre lógica algebraica en el siglo XIX sentó las bases de la lógica booleana. Esta se utiliza en computación y electrónica para expresar alternativas como verdadero/falso, sí/no o encendido/apagado, que también pueden representarse mediante `1` y `0`.

### Operadores lógicos

   Se utilizan para combinar o negar condiciones. Cuando los operandos son booleanos:

   * **`and` — «y»**: produce `True` si ambas condiciones son verdaderas y `False` en cualquier otro caso.
   * **`or` — «o»**: produce `True` si al menos una condición es verdadera; solo produce `False` cuando ambas son falsas.
   * **`not` — «no»**: invierte el valor lógico, de modo que `not True` es `False` y `not False` es `True`.

   Por ejemplo, `temperatura > 30 and humedad > 60` comprueba que se cumplan **las dos condiciones**. Estas operaciones permiten tomar decisiones en estructuras como `if`, `elif` y `else`.

<p style="margin: 0.45em 0;">No confundáis <code>=</code> con <code>==</code>. El primero <strong>asigna</strong> un valor y el segundo <strong>compara</strong> dos valores. En su uso general, <code>and</code> y <code>or</code> devuelven uno de sus operandos, que no tiene por qué ser booleano; <code>not</code> siempre devuelve <code>True</code> o <code>False</code>.</p>

## Tipos de valores en Python

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">El <strong>tipo</strong> de un valor indica qué clase de dato representa y qué operaciones podemos realizar con él. Para trabajar con números utilizaremos principalmente <strong>enteros</strong> (<code>int</code>) y <strong>flotantes</strong> (<code>float</code>); más adelante veremos también los complejos (<code>complex</code>).</p>
</div>

Los **enteros (`int`)** representan valores como `5`, `-3` o `0`. Se almacenan de forma exacta, sin redondeo, y su tamaño puede crecer hasta los límites que imponga la memoria disponible.

   Los **flotantes (`float`)** permiten representar números con parte decimal, como `-3.2`, pero también valores como `5.0`. Su precisión es finita, por lo que algunos números no se representan exactamente y pueden aparecer pequeñas diferencias de redondeo. Por ejemplo, `1 / 3` produce una aproximación como `0.3333333333333333`.

   Al sumar, restar o multiplicar un `int` y un `float`, normalmente obtenemos un `float`. Python convierte el entero para realizar la operación con el tipo flotante; esto no supone una ganancia general de precisión. La división con `/` también produce un `float` cuando ambos operandos son enteros, incluso si el cociente es un número entero.

   Podemos convertir explícitamente entre estos tipos con `int()` y `float()`. Por ejemplo, `float(3)` produce `3.0`. En cambio, `int(3.9)` produce `3` e `int(-3.9)` produce `-3`, porque **la conversión a entero trunca hacia cero, no redondea al entero más cercano**.

   Conocer el tipo resulta importante al comparar valores, realizar cálculos o guardar resultados en una base de datos. Tampoco debemos deducir el consumo de memoria solo por la presencia de decimales, ya que depende de la implementación y, en los enteros, del tamaño del número.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #C18A36; background-color: rgba(193,138,54,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Un <code>float</code> no es siempre «más preciso» que un <code>int</code>. Los enteros son exactos; los flotantes ofrecen un rango amplio de valores, pero con precisión finita.</p>
</div>

Dos cálculos matemáticamente equivalentes pueden producir flotantes ligeramente distintos. Para comprobar una igualdad aproximada podemos admitir una pequeña diferencia, como en `abs(a - b) < 1e-9`. La función `abs()` obtiene el valor absoluto, de modo que aquí comparamos el tamaño de la diferencia sin importar su signo.

   La tolerancia debe elegirse según la escala de los valores, sus unidades y la precisión necesaria. `1e-9` no es un umbral universal. Observad qué ocurre al sumar `0.1` y `0.2`.

````python
a = 0.1 + 0.2
b = 0.3

print(a)                    # La representación no es exactamente 0.3.
print(a == b)               # Comparación exacta: False.
print(abs(a - b) < 1e-9)     # Comparación con tolerancia: True.
````

## Operadores aritméticos en Python

Veamos ahora los operadores aritméticos en acción. En cada ejemplo calculamos un resultado, lo guardamos en una variable y lo mostramos con `print()`.

   Fijaos especialmente en la diferencia entre `/`, `//` y `%`, que representan la división, el cociente redondeado hacia abajo y el resto, respectivamente.

````python
# Suma: añadimos el segundo número al primero.
suma = 3 + 2
print(suma)
````

````python
# Multiplicación de dos números.
multiplicacion = 3 * 2
print(multiplicacion)
````

````python
# División entera: redondeamos el cociente hacia menos infinito.
# Con dos operandos enteros, el resultado también es un int.
division_entera = -3 // 2.5
print(division_entera)
````

````python
# Resto: 3 = 1 * 2 + 1.
modulo = 3 % 2
print(modulo)
````

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #C18A36; background-color: rgba(193,138,54,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Con números positivos, <code>//</code> puede parecer equivalente a quitar los decimales. Con números negativos se aprecia la diferencia. Por ejemplo, <code>-3 // 2</code> da <code>-2</code>, no <code>-1</code>, porque se redondea <strong>hacia menos infinito</strong>. En cambio, <code>int(-1.5)</code> da <code>-1</code>, porque trunca hacia cero.</p>
</div>

La división entera y el módulo están relacionados. Para enteros y un divisor `b` distinto de cero, se cumple la siguiente igualdad.

   ```python
   a == (a // b) * b + (a % b)
   ```

   Por ejemplo, `-3 = (-2) * 2 + 1`. Además, «división entera» describe la operación, no garantiza que el tipo del resultado sea `int`. Si un operando es flotante, podemos obtener un `float`.

````python
print(-3 // 2)   # -2: redondeo hacia menos infinito.
print(-3 % 2)    # 1: se cumple -3 = (-2) * 2 + 1.
print(3.0 // 2)  # 1.0: un operando es float.
````

````python
# Potenciación: elevamos la base 5 al exponente 2.
potencia = 5 ** 2
print(potencia)
````

## Números complejos

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Un <strong>número complejo</strong> tiene una parte real y una parte imaginaria. Python dispone del tipo <code>complex</code> y utiliza la notación <code>j</code> para escribir la parte imaginaria. Por ejemplo, <code>3 + 4j</code> representa un número con parte real <code>3</code> y parte imaginaria <code>4</code>.</p>
</div>

Para escribir la unidad imaginaria utilizamos `1j`, ya que `j` por sí solo sería un nombre de variable, no una constante especial.

   El **conjugado** conserva la parte real y cambia el signo de la imaginaria. El producto de un complejo por su conjugado es el cuadrado de su módulo. Para $x = 3 + 4j$, obtenemos lo siguiente.

   $$
   x\overline{x} = 3^2 + 4^2 = 25.
   $$

   En Python veremos el producto como `(25+0j)`. Su valor es real, pero el resultado sigue siendo de tipo complejo.

````python
# Complejo con parte real 3 y parte imaginaria 4.
x = 3 + 4j
print(x)
````

````python
# El conjugado cambia el signo de la parte imaginaria.
print(x.conjugate())
````

````python
# El producto por el conjugado es el cuadrado del módulo.
print(x * x.conjugate())
````

<p style="margin: 0.45em 0;">La <strong>notación científica</strong> permite escribir de forma compacta números muy grandes o muy pequeños. En Python se utiliza <code>e</code> o <code>E</code> para indicar una potencia de diez.</p>
</div>

Por ejemplo, `1.5e9` representa $1.5 \times 10^9$, mientras que `1.5e-4` representa $1.5 \times 10^{-4}$.

   La `e` forma parte de la escritura del número y **no es la constante de Euler**. Un valor escrito con esta notación se interpreta como un `float`, aunque no tenga punto decimal.

````python
# 1.5 multiplicado por diez elevado a 9.
num_grande = 1.5e9
print(num_grande)
````

````python
# 1.5 multiplicado por diez elevado a -4.
num_pequeño = 1.5e-4
print(num_pequeño)
````

### Sublistas (*slicing*)

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Además de consultar un elemento, podemos seleccionar una parte de la lista. Esta operación se llama <strong>corte</strong> o <em><strong>slicing</strong></em> y se escribe como <code>lista[inicio:fin:paso]</code>. El inicio se incluye; el límite de parada, <strong>no</strong>.</p>
</div>

Esta notación también suele aparecer como `lista[i:j:l]`: `i` es el índice inicial, `j` el límite de parada y `l` el paso. Con paso positivo avanzamos hacia índices mayores; con paso negativo recorremos la lista en sentido contrario.

   Si no indicamos el paso, se utiliza `1`. En ese caso podemos omitir el segundo signo `:`, de modo que `lista[i:j]` equivale a `lista[i:j:1]`. Con paso positivo, omitir el inicio permite empezar por el primer elemento, y omitir el final permite llegar hasta el último.

   Para obtener las temperaturas **del cuarto al sexto día**, seleccionamos los índices `3`, `4` y `5`. Por eso escribimos `3:6`, ya que el índice `6` queda fuera del corte.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 3px solid #72A0C1; background-color: rgba(76,120,168,0.025); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0 0 0.4em 0; font-weight: 600;">El límite y el paso</p><p style="margin: 0.45em 0;">Con paso positivo se seleccionan índices menores que el límite de parada; con paso negativo, índices mayores que ese límite. La regla «termina en <code>j - 1</code>» solo describe el caso en que avanzamos de uno en uno y el corte llega hasta esa posición. Además, los límites de un corte se ajustan a la longitud disponible y no funcionan igual que el acceso a un elemento individual.</p>
</div>

````python
# Índices 3, 4 y 5: del cuarto al sexto día.
print(temperaturas[3:6])
````

````python
# Indicar paso 1 produce el mismo resultado.
print(temperaturas[3:6:1])
````

En ambos casos obtenemos `[16.3, 18.5, 19.7]`. El valor de índice `6`, correspondiente al séptimo día, no aparece porque el límite de parada está excluido.

   Para incluir ese último dato podemos usar `7` como límite. Si escribimos un límite mucho mayor, como `127`, el corte sigue llegando solo hasta el final de la lista.

````python
# Incluimos también el índice 6.
print(temperaturas[3:7])
````

````python
# El límite del corte se ajusta a los elementos disponibles.
print(temperaturas[3:127])
````

Con paso positivo, un límite superior demasiado grande no genera un error, porque el corte se detiene al final de la lista. Sin embargo, para expresar que queremos llegar hasta el último elemento resulta más claro **omitir el límite**.

````python
# Sin límite final: desde el índice 3 hasta el último elemento.
print(temperaturas[3::])
````

Si indicamos un paso de `2`, avanzamos de dos en dos. Desde el índice `3`, se seleccionan los índices `3` y `5`.

````python
print(temperaturas[3::2])
````

También podemos combinar un inicio positivo con un límite negativo. En `descripciones[0:-1:2]` empezamos por el primer elemento, dejamos fuera el último y avanzamos de dos en dos.

   Se seleccionan los índices `0`, `2` y `4`. El penúltimo elemento, de índice `5`, tampoco aparece, pero no porque lo excluya el límite, sino porque no se alcanza con ese paso.

````python
print(descripciones[0:-1:2])
````

Para empezar por el primer elemento y dejar fuera los dos últimos, omitimos el inicio y usamos `-2` como límite. Con paso `2`, este ejemplo selecciona los índices `0`, `2` y `4`, por lo que termina en el antepenúltimo elemento.

````python
print(temperaturas[:-2:2])
````

Si avanzamos de uno en uno, obtenemos todos los elementos hasta el antepenúltimo. Las dos formas siguientes son equivalentes, ya que en ambas, el paso predeterminado es `1`.

````python
print(temperaturas[:-2:])
````

````python
print(temperaturas[:-2])
````

Cuando el inicio y el final del corte coinciden, no se selecciona ningún elemento. El resultado es una **lista vacía**, representada por `[]`.

````python
print(descripciones[2:2])
````

El corte `[:]` selecciona todos los elementos y crea una nueva lista con ellos. Podemos utilizarlo para mostrar el contenido completo.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 3px solid #72A0C1; background-color: rgba(76,120,168,0.025); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0 0 0.4em 0; font-weight: 600;">Una precisión sobre las copias</p><p style="margin: 0.45em 0;">Se trata de una <strong>copia superficial</strong>: se crea una lista exterior nueva, pero no se duplican los objetos que contiene. Si uno de sus elementos fuera otra lista, esa lista interior seguiría compartida con el original.</p>
</div>

````python
print(temperaturas[:])
````

Con un paso negativo podemos recorrer la lista en sentido contrario. En `[::-1]`, Python toma los extremos adecuados para empezar por el último elemento y retroceder hasta el primero.

   El paso no puede ser cero, porque no permitiría avanzar de una posición a otra.

````python
# El corte devuelve otra lista; no cambia el orden de temperaturas.
print(temperaturas[::-1])
````

### Modificar listas

Las listas son **mutables**: podemos cambiar sus elementos después de crearlas. Supongamos que detectamos un error en la temperatura del segundo día y queremos sustituirla por `18.4` °C. Asignamos el nuevo valor a la posición correspondiente.

````python
# El segundo día ocupa la posición de índice 1.
temperaturas[1] = 18.4
print(temperaturas)
````

## Funciones y métodos incorporados en Python

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Una <strong>función</strong> reúne instrucciones que podemos reutilizar al llamarla. Algunas ya están incorporadas en Python, como <code>print()</code> y <code>len()</code>. Un <strong>método</strong> realiza una operación asociada a un objeto y se utiliza escribiendo un punto tras su nombre, como en <code>temperaturas.append(20.4)</code>.</p>
</div>

La diferencia se aprecia en la escritura. `len(temperaturas)` llama a una función y le pasa la lista como argumento; `temperaturas.append(20.4)` llama a un método de esa lista. Otros métodos, como `remove()`, permiten eliminar elementos.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #C18A36; background-color: rgba(193,138,54,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Los ejemplos siguientes modifican las listas anteriores. Al añadir, eliminar u ordenar elementos por separado, las temperaturas y las descripciones <strong>dejan de corresponder al mismo día</strong>. Aquí observamos el efecto de cada operación; no estamos conservando una tabla de observaciones emparejadas. Para repetir la secuencia desde el inicio, volved a ejecutar las celdas que crean las listas.</p>
</div>

### Métodos comunes de las listas

**`append()` — añadir al final.** Incorpora un elemento al final de la lista. Por ejemplo, podemos añadir una temperatura de `20.4` °C.

````python
# Añadimos un dato y observamos la lista modificada.
temperaturas.append(20.4)
print(temperaturas)
````

**`remove()` — eliminar por valor.** Borra la **primera aparición** del valor indicado, no todas las coincidencias. En este ejemplo eliminamos la descripción `"lluvioso"`.

   Si ese valor no está en la lista, se genera un `ValueError`. Por eso, repetir esta celda después de haber eliminado la única aparición produciría un error.

````python
descripciones.remove("lluvioso")
print(descripciones)
````

**`insert()` — insertar en una posición.** Añade un elemento antes del que ocupa el índice indicado. Los elementos situados a partir de esa posición se desplazan una posición hacia la derecha, por lo que **no se sustituye** el que ya estaba allí.

````python
# Insertamos una descripción en la tercera posición.
descripciones.insert(2, "tormenta")
print(descripciones)
````

**`pop()` — extraer un elemento.** Elimina el elemento de la posición indicada y, además, **lo devuelve**, de modo que podemos guardarlo en una variable.

   Sin un índice, extrae el último elemento. Si la posición no existe, se genera un `IndexError`.

````python
# Guardamos el elemento que retiramos de la lista.
descripcion_eliminada = descripciones.pop(2)
print(descripcion_eliminada)
print(descripciones)
````

**`sort()` — ordenar la lista.** Ordena sus elementos en el propio objeto. Por defecto, el orden es ascendente. Funciona con números y también con otros elementos que puedan compararse entre sí, como cadenas de texto.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #C18A36; background-color: rgba(193,138,54,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;"><code>sort()</code> modifica la lista y devuelve <code>None</code>, <strong>no una lista nueva</strong>. Por eso escribimos <code>temperaturas.sort()</code>, sin asignar ese resultado a <code>temperaturas</code>. Además, al ordenar las temperaturas perdemos el orden cronológico de los días.</p>
</div>

````python
# Ordenamos los valores de menor a mayor, no por fecha.
temperaturas.sort()
print(temperaturas)
````

**`index()` — localizar un valor.** Devuelve el índice de su primera aparición. El valor buscado no tiene por qué ser numérico; si no se encuentra, se genera un `ValueError`.

   Ahora buscamos `18.5` en la lista **ya ordenada**, por lo que el índice obtenido no representa su posición en la semana original.

````python
indice = temperaturas.index(18.5)
print(indice)
````

**`len()` — contar elementos.** A diferencia de los ejemplos anteriores, es una **función**, no un método. Recibe la lista entre paréntesis y devuelve su longitud.

````python
dias_registrados = len(temperaturas)
print(dias_registrados)
````

La función incorporada **`type()`** nos permite consultar el tipo del objeto. En este caso confirma que `temperaturas` sigue siendo una lista.

````python
type(temperaturas)
````

## Tuplas en Python

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">

Las tuplas resultan útiles para agrupar valores que deben mantenerse juntos sin cambiar la composición de la colección. Habitualmente se escriben entre paréntesis `()`, con los elementos separados por comas.

<p style="margin: 0 0 0.4em 0; font-weight: 600;">Dos matices importantes</p><p style="margin: 0.45em 0;">En una tupla no vacía, es la <strong>coma</strong> la que define la agrupación. <p style="margin: 0.45em 0;">La inmutabilidad afecta a la tupla, no necesariamente a los objetos que contiene, de modo que una lista dentro de una tupla podría seguir modificándose.</p>

## Diccionarios en Python

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Un <strong>diccionario</strong> almacena pares de <strong>clave y valor</strong>. La clave identifica el dato que queremos consultar. Por ejemplo, la clave <code>"ciudad"</code> puede estar asociada al valor <code>"Barcelona"</code>. Cada clave es única dentro del diccionario.</p>
</div>

La idea se parece a la de un diccionario de palabras. Buscamos una palabra, la clave, y obtenemos su definición, el valor. A diferencia de una lista, no necesitamos conocer la posición del dato, sino su clave.

   ### Creación de diccionarios

   Los pares se escriben entre llaves `{}`, separados por comas. Dentro de cada par, separamos la clave y el valor con dos puntos, como en `clave: valor`. Crearemos un registro ficticio con temperatura en **°C**, humedad relativa en **%** y presión en **hPa**.

````python
# Registro meteorológico ficticio de una ciudad.
datos_meteorologicos = {
    "ciudad": "Barcelona",
    "temperatura": 22.5,  # °C
    "humedad": 65,       # %
    "presion": 1012,     # hPa
}
````

### Acceso a los valores

Escribimos la clave entre corchetes para recuperar el valor asociado. Por ejemplo, consultamos el nombre de la ciudad.

   Si la clave no existe, este acceso genera un `KeyError`.

````python
print(datos_meteorologicos["ciudad"])
````

### Modificación y adición

Los diccionarios son **mutables**. Una asignación a una clave existente sustituye su valor; si la clave es nueva, se añade el par al diccionario.

````python
# Actualizamos un valor existente.
datos_meteorologicos["temperatura"] = 23.0

# Añadimos una clave que todavía no estaba en el diccionario.
datos_meteorologicos["viento"] = "N 5 m/s"
print(datos_meteorologicos)
````

### Eliminación de claves

La sentencia `del` elimina la clave indicada y su valor asociado. En este caso retiramos la presión del registro.

````python
del datos_meteorologicos["presion"]
print(datos_meteorologicos)
````

### Métodos útiles

Podemos consultar por separado las claves, los valores o los pares completos:

   * **`keys()`** devuelve una vista de las claves.
   * **`values()`** devuelve una vista de los valores.
   * **`items()`** devuelve una vista de los pares clave–valor, representados como tuplas.

   Una **vista** permite consultar el contenido actual del diccionario sin crear una lista independiente. Si el diccionario cambia, la vista refleja ese cambio. Para obtener una lista de las claves en ese momento, podemos escribir `list(datos_meteorologicos.keys())`.

   En el bucle siguiente, `clave` y `valor` reciben los dos componentes de cada par, y `print()` los muestra juntos.

````python
for clave, valor in datos_meteorologicos.items():
    print(f"{clave}: {valor}")
````

````python
# Vista de las claves disponibles.
datos_meteorologicos.keys()
````

````python
# Vista de los valores asociados.
datos_meteorologicos.values()
````

El método **`get()`** también permite consultar una clave. La diferencia respecto a los corchetes es que, si no existe, devuelve `None` en lugar de generar un error.

   Podemos proporcionar un valor alternativo. Por ejemplo, `datos_meteorologicos.get("precipitacion", "sin datos")` devuelve `"sin datos"` si la clave no está presente. Esto distingue un dato **no disponible** de una precipitación observada de cero.

````python
# Una clave existente devuelve su valor, igual que con corchetes.
datos_meteorologicos.get("temperatura")
````

````python
print(datos_meteorologicos.get("ciudad"))
print(datos_meteorologicos.get("precipitacion"))             # None
print(datos_meteorologicos.get("precipitacion", "sin datos"))
````

## Trabajar con fechas en Python

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">En meteorología y climatología, un dato suele estar asociado a <strong>cuándo se observó o a qué momento se refiere</strong>. Trabajar con fechas y horas nos permite ordenar registros, seleccionar periodos y calcular duraciones.</p>
</div>

Las series históricas de temperatura, precipitación o viento suelen estar indexadas por fechas y horas. Las predicciones meteorológicas y las proyecciones climáticas también se relacionan con momentos o periodos determinados. Las marcas temporales, llamadas *timestamps*, permiten identificar esas referencias en los datos.

   Para trabajar con ellas, utilizaremos objetos específicos de fechas y horas, en lugar de tratarlas únicamente como texto.

### Módulo `datetime`

Python incluye **`datetime`** en su biblioteca estándar, por lo que no necesitamos instalar un paquete adicional para utilizarlo. Un **módulo** reúne herramientas relacionadas que podemos incorporar con `import`.

   `datetime` proporciona tipos de objetos y operaciones para trabajar con fechas, horas y duraciones. Empezamos por importarlo.

````python
import datetime
````

La función `dir()` permite explorar los nombres disponibles dentro de un módulo o un objeto. No es necesario conocerlos todos. Aquí sirve para ver que una biblioteca agrupa distintas herramientas.

   En `datetime.datetime`, el primer nombre es el **módulo** y el segundo, una **clase** definida dentro de él. Una clase describe un tipo de objeto; en este caso, uno que combina fecha y hora. Comparad lo que aparece al consultar ambos.

````python
# Nombres disponibles en el módulo.
print(dir(datetime))

# Nombres disponibles en la clase que combina fecha y hora.
dir(datetime.datetime)
````

### Creación de fechas

Con `datetime.date(año, mes, día)` creamos un objeto que representa una fecha concreta. Los tres argumentos indican, en ese orden, el año, el mes y el día.

La diferencia entre dos fechas, o entre dos objetos `datetime`, se representa mediante un **`datetime.timedelta`**: un objeto que expresa una duración.

### Operaciones con fechas

Podemos restar dos fechas para obtener el tiempo transcurrido entre ellas. En el ejemplo siguiente calculamos el intervalo desde el 1 de enero hasta el 31 de diciembre de 2023.

````python
fecha_inicio = datetime.date(2023, 1, 1)
fecha_fin = datetime.date(2023, 12, 31)

diferencia = fecha_fin - fecha_inicio
print(diferencia.days)  # Días transcurridos entre las dos fechas.
````

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #59A14F; background-color: rgba(89,161,79,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0 0 0.4em 0; font-weight: 600;">Interpretar el resultado</p><p style="margin: 0.45em 0;">¿Por qué obtenemos <code>364</code> y no <code>365</code>? La resta mide el <strong>tiempo transcurrido entre las fechas</strong>. Si queremos contar todas las fechas del calendario, incluyendo tanto el 1 de enero como el 31 de diciembre, debemos añadir uno. Así, <code>diferencia.days + 1</code> da <code>365</code>. Medir una duración y contar fechas incluidas en un periodo son preguntas distintas.</p>
</div>

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #C18A36; background-color: rgba(193,138,54,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0 0 0.4em 0; font-weight: 600;">Ejercicio</p><p style="margin: 0.45em 0;">Calculad la duración en días y segundos de un huracán hipotético que se inició el <strong>7 de septiembre de 2021</strong> y terminó el <strong>16 de septiembre de 2021</strong>.</p>
<p style="margin: 0.45em 0;">Tomamos ambas fechas a la misma hora y usamos días de 24 horas. Calculamos el intervalo transcurrido, no el número de fechas del calendario contando ambos extremos. Antes de ejecutar la solución de la celda siguiente, anticipad el resultado.</p>
</div>

````python
dia_inicio = datetime.date(2021, 9, 7)
dia_fin = datetime.date(2021, 9, 16)
duracion_huracan = dia_fin - dia_inicio

# Mostramos la duración expresada en dos unidades distintas.
print("Duración en días:", duracion_huracan.days)
print("Duración en segundos:", duracion_huracan.total_seconds())
````

En un intervalo positivo, el atributo **`.days`** indica los días completos y el método **`.total_seconds()`** expresa toda la duración en segundos. La diferencia en la escritura también importa, porque accedemos al atributo sin paréntesis y llamamos al método con `()`.

   No confundáis `.total_seconds()` con **`.seconds`**, que contiene solo los segundos restantes después de separar los días completos. Veamos un intervalo de dos días y tres horas.

````python
intervalo = datetime.timedelta(days=2, hours=3)

print(intervalo.days)             # 2 días completos.
print(intervalo.seconds)          # 10800 s: las 3 horas restantes.
print(intervalo.total_seconds())  # 183600.0 s: la duración completa.
````

### Intervalos de calendario con `relativedelta`

Los objetos `timedelta` expresan duraciones en días, segundos y microsegundos. Pero a veces nos interesa describir un intervalo en **años, meses y días de calendario**, por ejemplo, al comparar fechas de distintos periodos climáticos.

   Para ello utilizaremos la clase **`relativedelta`**, del módulo `dateutil.relativedelta`.

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #59A14F; background-color: rgba(89,161,79,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Los meses y los años no tienen siempre la misma duración. Por eso, <strong>sumar un mes de calendario no equivale a sumar un número fijo de días o segundos</strong>, con independencia de la fecha de partida.</p>
</div>

Esta herramienta pertenece a una biblioteca externa. Su nombre al importarla es `dateutil`, pero el paquete que se instala se llama **`python-dateutil`**.

   Si todavía no está instalado, utilizad **una** de estas alternativas en una terminal con el entorno de Python adecuado activado:

   ```bash
   conda install python-dateutil
   ```

   ```bash
   pip install python-dateutil
   ```

   No es necesario ejecutar ambas órdenes. Si el paquete ya está disponible en el entorno del kernel, podéis pasar directamente a la importación.

Importamos `relativedelta`. Repetimos también la importación de `datetime` para que quede claro qué herramientas utiliza este apartado.

````python
import datetime
from dateutil.relativedelta import relativedelta
````

Calcularemos la duración de un episodio **hipotético** de El Niño que comienza el 15 de enero de 2010 y termina el 25 de junio de 2011. Las fechas son ficticias y se utilizan únicamente para practicar el cálculo de intervalos.

````python
inicio_nino = datetime.date(2010, 1, 15)
fin_nino = datetime.date(2011, 6, 25)
````

Pasamos primero la fecha final y después la inicial. El resultado permite consultar los años, meses y días de calendario que separan ambas fechas.

````python
duracion = relativedelta(fin_nino, inicio_nino)

print(
    f"El episodio hipotético de El Niño duró {duracion.years} año, "
    f"{duracion.months} meses y {duracion.days} días."
)
````

Además de comparar fechas, `relativedelta` permite sumar o restar intervalos de calendario. Para obtener la fecha de hace **1 año, 2 meses y 3 días**, tomando como referencia la variable `hoy` que hemos creado antes, utilizamos la siguiente expresión.

````python
fecha_anterior = hoy - relativedelta(years=1, months=2, days=3)
print(fecha_anterior)
````

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 3px solid #72A0C1; background-color: rgba(76,120,168,0.025); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0 0 0.4em 0; font-weight: 600;">Al llegar al final de un mes</p><p style="margin: 0.45em 0;">Si el mes de destino no tiene el día de partida, <code>relativedelta</code> utiliza su último día. Por ejemplo, un mes después del <strong>31 de enero de 2024</strong> es el <strong>29 de febrero de 2024</strong>, porque febrero no tiene día 31 y, en ese año bisiesto, termina el día 29.</p>
</div>

````python
fecha_ejemplo = datetime.date(2024, 1, 31)
print(fecha_ejemplo + relativedelta(months=1))  # 2024-02-29
````

## Breve apunte sobre formato de cadenas de texto

<div style="padding: 0.8em 1em; margin: 0.8em 0 1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px; line-height: 1.6;">
<p style="margin: 0.45em 0;">Dar <strong>formato a una cadena de texto</strong> permite combinar palabras y valores, controlar su alineación o elegir cuántos decimales se muestran. La operación produce texto y no modifica por sí misma los datos numéricos originales.</p>
</div>

Nos centraremos en dos formas de hacerlo: las **f-strings** y el método **`str.format()`**.

   ### F-strings

   Una **f-string** es una cadena precedida de `f` o `F`, disponible desde Python 3.6. Dentro de las comillas, las llaves `{}` permiten insertar el valor de una variable o el resultado de una expresión.

**Insertar una variable.** El contenido de `{nombre}` se sustituye por el valor de la variable al crear la cadena.

````python
nombre = "Alice"
saludo = f"Hola, {nombre}"
print(saludo)
````

**Alineación y ancho del campo.** Después de dos puntos podemos indicar cómo presentar el valor. `<` alinea a la izquierda, `>` a la derecha y `^` centra el contenido.

   En este ejemplo reservamos un ancho **mínimo** de 20 caracteres, que incluye el texto y los espacios de relleno. Si el texto fuera más largo, no se recortaría. Las barras `|` permiten ver dónde empieza y termina cada campo.

````python
nombre = "Alice"
alineado_izquierda = f"{nombre:<20}"  # Ancho mínimo de 20 caracteres.
alineado_derecha = f"{nombre:>20}"
centrado = f"{nombre:^20}"

# Las barras hacen visibles los espacios de relleno.
print(f"|{alineado_izquierda}|")
print(f"|{alineado_derecha}|")
print(f"|{centrado}|")
````

**Número de decimales.** El especificador `.2f` presenta un número en formato decimal fijo con dos cifras después del punto. Se redondea la representación que se muestra, no el valor almacenado en la variable original.

````python
pi = 3.14159
dos_decimales = f"{pi:.2f}"
print(dos_decimales)  # Texto: 3.14.
print(pi)            # El valor numérico original no ha cambiado.
````

**Evaluación de expresiones.** Entre las llaves también podemos escribir una operación. Python la evalúa e incorpora su resultado al texto.

````python
suma = f"{3 + 2}"
print(suma)       # Se muestra 5.
print(type(suma))  # La variable contiene texto, no el entero 5.
````

### `str.format()`

El método **`str.format()`**, disponible desde Python 2.6, permite dar formato a una cadena utilizando llaves `{}` como marcadores de posición. Los valores que escribimos entre los paréntesis de `.format()` se insertan en esos espacios.

   A diferencia de las f-strings, aquí la cadena no lleva una `f` delante.

````python
"Hola, {}".format(nombre)
````

**Insertar una variable.** Podemos guardar el resultado en otra variable, igual que con las f-strings.

````python
nombre = "Alice"
saludo = "Hola, {}".format(nombre)
print(saludo)
````

**Alineación y ancho del campo.** Los símbolos tienen el mismo significado que en las f-strings. Aquí reservamos un mínimo de diez caracteres y alineamos el nombre a la izquierda.

````python
nombre = "Alice"
alineado_izquierda = "{:<10}".format(nombre)
print("|{}|".format(alineado_izquierda))
````

**Número de decimales.** `.2f` conserva el mismo significado. El texto resultante muestra dos decimales, sin cambiar el valor numérico de `pi`.

````python
pi = 3.14159
dos_decimales = "{:.2f}".format(pi)
print(dos_decimales)
print(pi)  # La variable numérica sigue conservando su valor.
````
