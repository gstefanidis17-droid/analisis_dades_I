Una <strong>biblioteca</strong>, también llamada <strong>librería</strong>, reúne código que podemos reutilizar para resolver tareas sin escribirlo todo desde cero. Puede incluir funciones, clases y otros recursos para trabajar con números, texto, fechas o datos científicos.

## Diferencias entre módulo, paquete y biblioteca

Estos términos están relacionados, pero describen aspectos distintos de la organización del código.

<div style="padding: 0.8em 1em; margin: 0.9em 0 1.1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px;">
<strong>Módulo.</strong> Es una unidad de código que puede importarse. En su forma más habitual, es un archivo con extensión <code>.py</code> que contiene funciones, clases, variables u otras instrucciones.
</div>

Un archivo llamado `matematicas.py` puede reunir dos operaciones sencillas.

```python
def suma(a, b):
    """Devuelve la suma de dos números."""
    return a + b


def resta(a, b):
    """Devuelve la diferencia entre dos números."""
    return a - b
```

Para utilizarlo desde otro programa o desde un notebook, el archivo debe estar en una carpeta donde Python pueda encontrarlo. En este ejemplo lo situamos en el **directorio de trabajo**. Al importarlo, escribimos su nombre sin la extensión `.py`.

```python
import matematicas

resultado = matematicas.suma(3, 4)
print(resultado)  # Muestra 7.
```

No todos los módulos son archivos `.py`. También existen módulos integrados en el intérprete o implementados mediante código compilado, como ocurre con `math` en CPython, la implementación habitual de Python.

<div style="padding: 0.8em 1em; margin: 0.9em 0 1.1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px;">
<strong>Paquete.</strong> Es un módulo que permite organizar otros módulos y subpaquetes bajo un nombre común. Un <strong>paquete regular</strong> suele corresponder a una carpeta que contiene un archivo <code>__init__.py</code>.
</div>

La siguiente estructura mantiene un módulo independiente y agrupa otros dos dentro de `paquete_personal`.

```text
proyecto/
    llibreries_python_2026_2027.ipynb
    matematicas.py
    paquete_personal/
        __init__.py
        matematicas.py
        geometria.py
```

El archivo `paquete_personal/matematicas.py` contiene las mismas funciones `suma()` y `resta()` del ejemplo anterior. `geometria.py` representa otro módulo del paquete. El archivo `__init__.py` puede estar vacío o contener instrucciones de inicialización, que se ejecutan cuando el paquete se carga por primera vez en la sesión.

```python
from paquete_personal import matematicas

resultado = matematicas.suma(3, 4)
print(resultado)  # Muestra 7.
```

Aquí el nombre `matematicas` se refiere al módulo **dentro de `paquete_personal`**, no al archivo independiente situado junto al notebook. Si ejecutamos los dos ejemplos en la misma sesión, la segunda importación sustituye la referencia anterior asociada a ese nombre.

Existen también *paquetes de espacio de nombres*, que no necesitan `__init__.py`. En este cuaderno trabajamos con paquetes regulares para que su estructura sea explícita.

<div style="padding: 0.8em 1em; margin: 0.9em 0 1.1em; border-left: 4px solid #4C78A8; background-color: rgba(76,120,168,0.08); border-radius: 4px;">
<strong>Biblioteca o librería.</strong> Describe un conjunto de recursos reutilizables, no un formato de archivo ni un tipo adicional de importación. Una biblioteca puede distribuirse como un módulo, como un paquete o como varios paquetes relacionados.
</div>

Python incluye una **biblioteca estándar**, con recursos como `math` para operaciones matemáticas y `datetime` para fechas y horas. Las bibliotecas de terceros amplían estas posibilidades y deben estar instaladas en el entorno antes de utilizarlas.

## Instalación de bibliotecas, paquetes y módulos

**Instalar** y **importar** son operaciones diferentes. La instalación incorpora un paquete y sus dependencias al entorno de trabajo. La importación permite utilizar sus recursos en una sesión de Python. El módulo `math` ya forma parte de la biblioteca estándar y no necesita una instalación adicional.

Un **entorno** reúne un intérprete de Python y un conjunto de paquetes instalados. Por eso, una biblioteca puede estar disponible en un entorno y no en otro.

Si utilizamos conda, primero activamos el entorno del curso en una **terminal**. En la siguiente orden, `nombre_del_entorno` debe sustituirse por su nombre real.

```bash
conda activate nombre_del_entorno
```

Después podemos instalar, por ejemplo, NumPy desde los canales configurados.

```bash
conda install numpy
```

La opción `-c` permite indicar un canal, es decir, un repositorio de paquetes. Esta variante utiliza `conda-forge`.

```bash
conda install -c conda-forge numpy
```

Con pip, la orden equivalente para instalar NumPy en el entorno del intérprete seleccionado es la siguiente.

```bash
python -m pip install numpy
```

La forma `python -m pip` ejecuta pip con ese intérprete de Python y ayuda a evitar confusiones cuando hay varias instalaciones. Si el entorno utiliza otro nombre de comando, debe emplearse el que corresponda a su intérprete.

En una **celda de código de JupyterLab con kernel de IPython**, puede utilizarse `%pip` para dirigir la instalación al entorno de ese kernel.

```ipython
%pip install numpy
```

El símbolo `%` identifica una instrucción propia de IPython. No es sintaxis ordinaria de Python y no se copia tal cual en un archivo `.py`.

<div style="padding: 0.8em 1em; margin: 0.9em 0 1.1em; border-left: 4px solid #C58A32; background-color: rgba(197,138,50,0.09); border-radius: 4px;">
<strong>Elegir el entorno adecuado.</strong> Las órdenes anteriores son alternativas, no pasos que deban ejecutarse todos. En un entorno conda, utiliza primero conda y recurre a pip para lo que no esté disponible por esa vía. Evita alternar instalaciones sin controlar las dependencias. Si una biblioteca parece instalada pero no se puede importar, comprueba qué entorno utiliza el kernel del notebook.
</div>

Podemos consultar el intérprete que utiliza el kernel ejecutando este fragmento en una celda. La ruta ayuda a identificar el entorno donde deben estar instaladas las bibliotecas.

```python
import sys

print(sys.executable)
```

Los gestores intentan resolver e instalar las **dependencias declaradas** que sean necesarias. Esto no implica que incorporen todos los complementos opcionales ni que cualquier combinación de versiones sea compatible.

Además, el nombre de instalación y el de importación no siempre coinciden. Conviene consultar la documentación del proyecto en lugar de deducir uno a partir del otro.

## ¿Cómo se importan módulos, paquetes y bibliotecas?

Una **importación** permite utilizar código definido en otro módulo. La instrucción `import math`, por ejemplo, hace que el nombre `math` permita acceder a ese módulo dentro de nuestra sesión.

Al cargar un módulo por primera vez, Python ejecuta sus instrucciones de nivel superior, como las definiciones de funciones y clases. El cuerpo de una función no se ejecuta por el mero hecho de definirla. Normalmente, las importaciones posteriores reutilizan el módulo que ya está cargado.

Para facilitar la lectura, las importaciones suelen agruparse al inicio del programa. Primero se colocan las de la biblioteca estándar, después las de terceros y, finalmente, las del proyecto, separando los grupos con una línea en blanco.

### Importación de módulos

El siguiente ejemplo importa `math` y utiliza su función `sqrt()` para calcular una raíz cuadrada.

<p style="font-size: 0.9em; margin: 0.8em 0 0.3em;"><a href="https://docs.python.org/3/tutorial/modules.html">Python · Módulos</a> · <a href="https://peps.python.org/pep-0008/#imports">PEP 8 · Organización de las importaciones</a></p>

````python
import math

math.sqrt(16)
````

La **notación de punto** permite acceder a un recurso a través del objeto que lo contiene. En `math.sqrt(16)`, `math` es el módulo, `sqrt` es la función y `16` es el argumento con el que la llamamos.

Para explorar los nombres disponibles podemos usar `dir(math)`. La lista incluye funciones como `sqrt`, constantes como `pi` y nombres especiales como `__doc__`. No todos sus elementos son funciones ni métodos, y la lista puede variar según la versión de Python.

`dir()` sirve para orientarnos, pero no sustituye la documentación ni garantiza una descripción exhaustiva de cualquier objeto.

````python
# Consultamos los nombres disponibles, no el resultado de llamar a sus funciones.
print(dir(math))
````

Para consultar qué hace una función y cómo se utiliza, podemos pasarla a `help()`. Escribimos `help(math.sqrt)` sin llamar antes a `sqrt`, porque queremos obtener información sobre la función, no sobre un resultado numérico.

En la ayuda, `sqrt(x, /)` indica que `x` se proporciona por posición. Así, `math.sqrt(16)` es una llamada válida. La barra `/` forma parte de la descripción de la firma, no es un signo de división en ese contexto.

<p style="font-size: 0.9em; margin: 0.8em 0 0.3em;"><a href="https://docs.python.org/3/library/functions.html">Python · dir() y help()</a></p>

````python
help(math.sqrt)
````

## Importación con alias

Un **alias** es otro nombre con el que nos referimos al mismo objeto. La palabra `as` permite elegirlo al importar. Por ejemplo, podemos usar `m` en lugar de `math`, aunque en este caso el nombre original ya es breve.

El alias no cambia el nombre del paquete instalado ni crea una copia del módulo. En bibliotecas científicas se utilizan convenciones habituales, como `np` para NumPy y `pd` para pandas, que ayudan a reconocer el código.

````python
import math as m

print(m.sqrt(16))
````

También podemos crear esa referencia después de importar el módulo mediante una asignación. Tras `m = math`, ambos nombres apuntan al **mismo módulo**. La expresión `m is math` permite comprobar esa identidad.

````python
import math

m = math
print(m.sqrt(16))
print(m is math)  # True confirma que ambos nombres se refieren al mismo objeto.
````

## Importación de funciones específicas

Con `from ... import ...` podemos incorporar directamente los nombres que necesitamos, como una función, una clase o una constante. En lugar de escribir siempre `math.sqrt`, podremos utilizar `sqrt`.

Esta forma también carga el módulo si todavía no está disponible. **Selecciona qué nombres se incorporan a nuestra sesión**, no una parte aislada del archivo que Python ejecutaría por separado.

````python
from math import sqrt
from math import sin, cos, exp
````

Después de la importación podemos llamar a `sqrt(16)` sin el prefijo `math.`. El resultado es el mismo que con `math.sqrt(16)`.

La diferencia está en los nombres disponibles. En una sesión nueva, `from math import sqrt` incorpora `sqrt`, pero no crea por sí sola un nombre `math`. En este notebook ya disponemos de `math` porque lo hemos importado en una celda anterior.

````python
sqrt(16)
````

La variante `from math import *` incorpora de una vez los nombres que el módulo expone para esa operación. Si define `__all__`, se utilizan los nombres indicados allí; en caso contrario, se incorporan los que no comienzan por un guion bajo.

<div style="padding: 0.8em 1em; margin: 0.9em 0 1.1em; border-left: 4px solid #C58A32; background-color: rgba(197,138,50,0.09); border-radius: 4px;">
<strong>Una forma que conviene reconocer, pero evitar en el código habitual.</strong> El asterisco dificulta saber de dónde procede cada nombre y puede sustituir otros que ya existían. Por ejemplo, <code>math</code> aporta su propia función <code>pow</code>. Es preferible escribir <code>import math</code> o importar explícitamente los recursos necesarios.
</div>

El siguiente ejemplo se conserva para observar la sintaxis, no como modelo de organización de un programa.

<p style="font-size: 0.9em; margin: 0.8em 0 0.3em;"><a href="https://peps.python.org/pep-0008/#imports">PEP 8 · Importaciones con asterisco</a></p>

````python
# Ejemplo de sintaxis. En el código habitual, preferimos importaciones explícitas.
from math import *
````

Una función también puede recibir un alias mediante una asignación. En el siguiente ejemplo, `seno` se refiere a la misma función que `math.sin`; al asignarla no añadimos paréntesis, porque todavía no queremos ejecutarla.

La función `sin()` recibe el ángulo en **radianes**. Aunque el seno de π es exactamente cero en matemáticas, `math.pi` es una aproximación numérica. Por eso, `seno(math.pi)` puede devolver un valor muy pequeño, del orden de `1e-16`, en lugar de cero exacto. No es un problema del alias.

<p style="font-size: 0.9em; margin: 0.8em 0 0.3em;"><a href="https://docs.python.org/es/3/tutorial/floatingpoint.html">Python · Aritmética de punto flotante</a></p>

````python
seno = math.sin
seno(math.pi)
````

### Importación de paquetes

La sintaxis sigue las mismas reglas, pero el nombre puede incluir varios niveles separados por puntos. Retomamos el paquete `paquete_personal` definido al principio.

```python
import paquete_personal
```

Esta instrucción importa el paquete y ejecuta su inicialización si todavía no se había cargado. **No importa automáticamente todos los archivos que contiene.** Con un `__init__.py` vacío, por ejemplo, el submódulo `geometria` no queda disponible solo por importar `paquete_personal`.

Para utilizar un submódulo podemos importarlo de manera explícita y mantener su nombre completo.

```python
import paquete_personal.matematicas

print(paquete_personal.matematicas.suma(3, 4))
```

Otra posibilidad consiste en incorporar el nombre del submódulo a nuestra sesión.

```python
from paquete_personal import matematicas

print(matematicas.suma(3, 4))
```

También podemos importar directamente una función definida dentro de él.

```python
from paquete_personal.matematicas import suma

print(suma(3, 4))
```

Las tres variantes calculan lo mismo. Lo que cambia es el nombre que utilizamos para acceder a la función. Si un paquete expone algunos submódulos o funciones desde `__init__.py`, estos pueden estar disponibles tras la importación inicial, pero esa decisión depende de cómo esté organizado.

### Importación de bibliotecas

Una biblioteca no utiliza una sintaxis distinta. Importamos los módulos o paquetes que proporciona, con los nombres que indique su documentación. Por ejemplo, NumPy suele importarse con el alias `np`.

```python
import numpy as np

# Son tres temperaturas ilustrativas en grados Celsius, no observaciones reales.
temperaturas = np.array([18.0, 20.0, 22.0])
print(np.mean(temperaturas))  # Muestra 20.0, también en grados Celsius.
```

Este ejemplo necesita NumPy instalado. `np.array()` organiza los valores en un array y `np.mean()` calcula su media. La importación no calcula por sí misma la media ni implica que todos los submódulos de la biblioteca se hayan cargado.

<div style="padding: 0.8em 1em; margin: 0.9em 0 1.1em; border-left: 4px solid #59A14F; background-color: rgba(89,161,79,0.08); border-radius: 4px;">
<strong>Idea clave.</strong> Elige la forma de importación según los nombres que necesitas y la claridad del código. <code>import modulo</code>, <code>import modulo as alias</code> y <code>from modulo import funcion</code> ofrecen maneras distintas de acceder a los mismos recursos.
</div>

## Bibliotecas para el análisis de datos meteorológicos y climáticos

Estas bibliotecas cubren tareas complementarias. La tabla relaciona cada una con su uso y una forma habitual de importación. **No es necesario importarlas todas en cada programa**, ni instalarlas para ejecutar las celdas de este cuaderno.

| Biblioteca | Para qué se utiliza | Ejemplo en meteorología y clima | Importación habitual |
|---|---|---|---|
| [NumPy](https://numpy.org/doc/stable/) | Arrays y cálculo numérico. | Operar con una malla de temperaturas o una serie de observaciones. | `import numpy as np` |

**xarray y netCDF4 no son equivalentes.** xarray ofrece estructuras etiquetadas y operaciones de análisis, mientras que netCDF4 proporciona acceso al formato de archivo. xarray puede utilizar netCDF4 u otros motores para leer y escribir datos.
